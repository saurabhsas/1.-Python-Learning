# 1. What is Python?

Python is a general-purpose programming language used for automation, web applications, data analysis, artificial intelligence, and many other tasks.

# 6. Operators

## 6.1 Arithmetic operators

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `7 + 2` gives `9` |
| `-` | Subtraction | `7 - 2` gives `5` |
| `*` | Multiplication | `7 * 2` gives `14` |

Floor division rounds the quotient down to the nearest integer value. With negative numbers, this means toward negative infinity.

````python
a = 17
b = 5

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Power:", 2 ** 4)

# Negative floor division is worth checking carefully.
print(-7 // 2)  # -4
````

A string (`str`) is a sequence of characters. Strings are **immutable**, which means you cannot change an individual character in place.

# 9. Loops

Loops repeat work. Use `for` to iterate through a sequence or a known range; use `while` when repetition depends on a condition.

## 9.1 The `for` loop and `range()`

`range(stop)` generates numbers from `0` up to (but not including) `stop`. `range(start, stop, step)` gives more control.

````python
# Prints 0, 1, 2, 3, 4.
for number in range(5):
    print(number)

# Prints 2, 4, 6, 8.
for number in range(2, 10, 2):
    print(number)
````

## 9.2 Accumulators

An accumulator stores a running result. It is commonly used for totals, counts, and products.

````python
numbers = [2, 4, 6, 8]
total = 0

for number in numbers:
    total += number  # Add the current number to the running total.

print("Total:", total)
````

## 9.3 The `while` loop

A `while` loop continues while its condition is true. Update the relevant state so the loop can stop; otherwise it may run forever.

````python
countdown = 3

while countdown > 0:
    print(countdown)
    countdown -= 1  # Move toward the stopping condition.

print("Go!")
````

## 9.4 `break` and `continue`

- `break` exits the nearest loop.
- `continue` skips the rest of the current iteration and moves to the next one.

````python
# Stop when we find the first number divisible by 7.
for number in range(1, 30):
    if number % 7 == 0:
        print("First multiple of 7:", number)
        break

# Print only odd numbers.
for number in range(1, 8):
    if number % 2 == 0:
        continue
    print(number)
````

## 9.5 Nested loops

A nested loop is a loop inside another loop. A simple nested loop can perform many operations: if both loops run `n` times, the total number of iterations is roughly `n × n`, or `O(n²)`.

````python
# Print every pair (row, column) in a 2-by-3 grid.
for row in range(2):
    for column in range(3):
        print(row, column)
````

# 10. Lists

A list stores an ordered collection of items. Lists are **mutable**, so their contents can be changed after creation. They can contain duplicate values and values of different types, though a consistent type is often easier to work with.

Lists are important in coding interviews because they are commonly used to represent arrays.

````python
numbers = [10, 20, 30, 20]

print(numbers)
print(numbers[0])   # First item
print(numbers[-1])  # Last item
print(numbers[1:3]) # Slice: index 1 up to (not including) 3

numbers[0] = 99     # Lists can be modified in place.
print(numbers)
````

## 10.1 Common list operations

- `append(x)`: add one item to the end.
- `extend(iterable)`: add each item from an iterable.
- `insert(i, x)`: insert at an index.
- `remove(x)`: remove the first matching value.
- `pop()`: remove and return the last item by default.
- `sort()`: sort the list in place.
- `sorted(items)`: return a new sorted list.

````python
items = [3, 1, 4]

items.append(2)
print(items)

items.extend([5, 6])
print(items)

items.insert(0, 99)
print(items)

last_item = items.pop()
print("Popped:", last_item)
print(items)

items.sort()
print("Sorted in place:", items)

new_sorted = sorted([8, 2, 5])
print("New sorted list:", new_sorted)
````

## 10.2 Looping through a list

Use `for item in items` when you need each value. Use `enumerate()` when you need both the index and value.

````python
fruits = ["apple", "banana", "orange"]

for fruit in fruits:
    print(fruit)

for index, fruit in enumerate(fruits):
    print(index, fruit)
````

## 10.3 List aliasing and copying

Assignment does not automatically copy a list. Two variables can refer to the same list. Use `copy()` or slicing for a shallow copy when appropriate.

````python
original = [1, 2, 3]
alias = original        # Both names refer to the same list.
copy_of_original = original.copy()

alias.append(4)

print("Original:", original)
print("Alias:", alias)
print("Copy:", copy_of_original)
````

## 10.4 Two-dimensional lists

A list can contain other lists. Access an item with two indexes, such as `grid[row][column]`.

````python
grid = [
    [1, 2, 3],
    [4, 5, 6]
]

print(grid[0][0])  # 1
print(grid[1][2])  # 6

for row in grid:
    print(row)
````

# 11. Tuples

A tuple is an ordered collection that cannot be changed in place (it is immutable). Tuples are useful for fixed groups of values and for returning multiple values from a function.

````python
point = (3, 7)

print(point[0])
print(point[1])

# Tuple unpacking assigns each element to a variable.
x, y = point
print("x =", x, "y =", y)

# A one-item tuple needs a trailing comma.
single_item_tuple = (5,)
print(type(single_item_tuple))
````

# 12. Sets

A set stores unique values. It is useful for removing duplicates and testing membership. Sets do not provide positional indexing.

Set operations include union (`|`), intersection (`&`), difference (`-`), and symmetric difference (`^`).

````python
numbers = [1, 2, 2, 3, 3, 3]
unique_numbers = set(numbers)

print(unique_numbers)
print(2 in unique_numbers)

first = {1, 2, 3}
second = {3, 4, 5}

print("Union:", first | second)
print("Intersection:", first & second)
print("Difference:", first - second)
print("Symmetric difference:", first ^ second)
````

## 12.1 Why sets matter for problem-solving

Checking membership in a set is typically average-case `O(1)`, while searching through a list is `O(n)`. This can make sets useful when checking many values, although exact performance depends on the situation.

````python
# Keep track of values already seen.
seen = set()
values = [4, 2, 4, 7, 2]

for value in values:
    if value in seen:
        print("Duplicate found:", value)
    else:
        seen.add(value)
````

# 13. Dictionaries

A dictionary stores key-value pairs. Keys are unique and must be hashable; common key types include strings, integers, and tuples of hashable values.

Dictionaries are useful for counting, looking up information, and mapping one value to another.

````python
student = {
    "name": "Meera",
    "age": 21,
    "course": "Python"
}

print(student["name"])
print(student.get("grade", "Not available"))

student["age"] = 22          # Update an existing key.
student["city"] = "Delhi"    # Add a new key.

print(student)
````

## 13.1 Looping through dictionaries

Use `.keys()`, `.values()`, or `.items()` depending on what you need. `.items()` gives key-value pairs.

````python
scores = {"Asha": 88, "Ravi": 92, "Meera": 85}

for name in scores:
    print(name)

for score in scores.values():
    print(score)

for name, score in scores.items():
    print(f"{name}: {score}")
````

## 13.2 Frequency counting pattern

Counting occurrences is one of the most useful dictionary patterns for coding interviews.

````python
letters = "banana"
frequency = {}

for letter in letters:
    # get(letter, 0) returns zero when this key has not appeared yet.
    frequency[letter] = frequency.get(letter, 0) + 1

print(frequency)
````

# 14. Functions

A function is a reusable block of code. Functions make programs easier to organize, test, and understand.

- `def` defines a function.
- Parameters are names in the function definition.
- Arguments are values passed to a function call.
- `return` sends a result back to the caller.

````python
def greet(name):
    # name is a parameter.
    return f"Hello, {name}!"

message = greet("Asha")  # "Asha" is an argument.
print(message)
````

## 14.1 Parameters, return values, and `None`

A function that reaches the end without a `return` statement returns `None`. `print()` displays a value; `return` gives a value back to the caller. They are not interchangeable.

````python
def add(a, b):
    # Return the calculated result so it can be reused.
    return a + b

def display_sum(a, b):
    # This function displays the result but does not return it.
    print(a + b)

result = add(3, 4)
print("Returned result:", result)

display_result = display_sum(3, 4)
print("Return from display_sum:", display_result)  # None
````

## 14.2 Default and keyword arguments

A parameter can have a default value. Keyword arguments make calls more readable.

````python
def describe_person(name, city="Unknown"):
    return f"{name} lives in {city}."

print(describe_person("Ravi"))
print(describe_person("Meera", city="Noida"))
````

## 14.3 Functions with lists

Be aware that lists are mutable. If a function modifies a list passed to it, the caller can observe that change.

````python
def add_item(items, item):
    # This mutates the original list.
    items.append(item)

shopping_list = ["milk", "bread"]
add_item(shopping_list, "apples")

print(shopping_list)
````

A variable created inside a function is normally local to that function. Prefer passing values as parameters and returning results instead of relying on global variables.

````python
def calculate_square(number):
    square = number * number  # Local variable.
    return square

print(calculate_square(5))
# print(square)  # Would raise NameError because square is local.
````

## 15.2 Recursion

Recursion happens when a function calls itself. Every recursive solution needs:
1. A **base case** that stops recursion.
2. A **recursive case** that makes progress toward the base case.

Recursion is useful for some algorithms, but deep recursion can exceed Python's recursion limit.

````python
def factorial(n):
    # Base case: factorial of 0 and 1 is 1.
    if n <= 1:
        return 1

    # Recursive case: reduce the problem size.
    return n * factorial(n - 1)

print(factorial(5))  # 5 * 4 * 3 * 2 * 1 = 120
````

# 16. Comprehensions

A comprehension builds a new collection from an iterable in a concise way. Learn the ordinary loop first; use comprehensions when they remain readable.

````python
# Build a list of squares.
squares = [number ** 2 for number in range(1, 6)]
print(squares)

# Build a list containing only even numbers.
evens = [number for number in range(1, 11) if number % 2 == 0]
print(evens)

# Build a dictionary mapping each number to its square.
square_map = {number: number ** 2 for number in range(1, 5)}
print(square_map)

# Build a set of unique word lengths.
word_lengths = {len(word) for word in ["cat", "dog", "python", "cat"]}
print(word_lengths)
````

# 17. Exceptions and debugging

An exception is an error detected while the program is running. Common examples include:
- `ValueError`: a value has the right general type but is not valid for the operation, such as `int("hello")`.
- `TypeError`: an operation is used with an unsuitable type.
- `IndexError`: a sequence index is out of range.
- `KeyError`: a dictionary key does not exist.
- `ZeroDivisionError`: division by zero.

Use `try` and `except` to handle errors you reasonably expect. Avoid catching every possible exception without a plan.

````python
text_value = "hello"

try:
    number = int(text_value)
    print(number)
except ValueError:
    # Handle the specific conversion problem.
    print("Please provide text that represents a whole number.")
````

## 17.1 Debugging habits

1. Read the full traceback, especially the final line.
2. Find the line where the error occurred.
3. Inspect variable values and types.
4. Reduce the problem to a small example.
5. Fix the cause rather than hiding the error.
6. Rerun the relevant test cases.

````python
# Example of a defensive check.
def safe_divide(a, b):
    if b == 0:
        return None  # A simple example policy; other programs may raise an error.
    return a / b

print(safe_divide(10, 2))
print(safe_divide(10, 0))
````

# 20. Object-oriented programming basics

Object-oriented programming (OOP) groups data and behavior together.

- A **class** is a blueprint.
- An **object** is an instance of a class.
- `__init__` initializes a new object.
- `self` refers to the current instance.
- An **attribute** stores data on an object.
- A **method** is a function defined inside a class.

````python
class Student:
    def __init__(self, name, score):
        # Store data on this particular object.
        self.name = name
        self.score = score

    def passed(self):
        # Return whether this student meets the example threshold.
        return self.score >= 50

student = Student("Asha", 82)

print(student.name)
print(student.score)
print(student.passed())
````

# 22. Testing and edge cases

A solution that works for one example may still be wrong. Test normal and boundary cases.

For algorithmic problems, consider:
- the smallest allowed input,
- empty input if permitted,
- one element,
- repeated values,
- negative numbers or zero if allowed,
- already sorted or reverse-sorted input,
- large inputs,
- and values at stated constraints.

````python
def average(numbers):
    # Explicitly reject an empty list so the behavior is clear.
    if len(numbers) == 0:
        raise ValueError("average() requires at least one number")
    return sum(numbers) / len(numbers)

# A small set of tests.
assert average([2, 4, 6]) == 4
assert average([5]) == 5

try:
    average([])
except ValueError as error:
    print("Expected error:", error)

print("Tests completed.")
````

# 26. Answer key — compare after attempting the exercises

Try to complete the practice cells first. The solutions below are examples; there may be other correct approaches.

````python
# 1. Celsius to Fahrenheit
celsius = 37
fahrenheit = celsius * 9 / 5 + 32
print(fahrenheit)

# 2. Divisible by both 3 and 5
number = 45
print(number % 3 == 0 and number % 5 == 0)

# 3. Multiplication table
number = 6
for multiplier in range(1, 11):
    print(f"{number} x {multiplier} = {number * multiplier}")

# 4. Factorial using a loop
n = 5
factorial_result = 1
for value in range(1, n + 1):
    factorial_result *= value
print(factorial_result)

# 5. Count vowels
text = "Learning Python"
vowels = "aeiou"
count = 0
for character in text.lower():
    if character in vowels:
        count += 1
print(count)

# 6. Reverse a string using a loop
text = "python"
reversed_text = ""
for character in text:
    reversed_text = character + reversed_text
print(reversed_text)

# 7. Find the smallest value without min()
numbers = [8, 3, 11, -2, 5]
smallest = numbers[0]
for value in numbers:
    if value < smallest:
        smallest = value
print(smallest)

# 8. Remove duplicates while preserving order
numbers = [3, 1, 3, 2, 1, 4]
unique_in_order = []
seen = set()
for value in numbers:
    if value not in seen:
        unique_in_order.append(value)
        seen.add(value)
print(unique_in_order)

# 9. Word frequency
sentence = "python is fun and python is useful"
word_counts = {}
for word in sentence.split():
    word_counts[word] = word_counts.get(word, 0) + 1
print(word_counts)

# 10. Second-largest distinct value
def second_largest_distinct(numbers):
    distinct_values = set(numbers)
    if len(distinct_values) < 2:
        return None
    ordered_values = sorted(distinct_values)
    return ordered_values[-2]

print(second_largest_distinct([5, 1, 5, 3]))  # 3
print(second_largest_distinct([4, 4]))        # None
print(second_largest_distinct([-5, -2, -9]))  # -5
````
