# 1. A repeatable problem-solving framework

Do not start coding the moment you read a problem. Use this sequence:

1. **Restate:** Explain the task in your own words.
2. **Clarify:** Identify inputs, outputs, constraints, and special cases.
3. **Example:** Work through a small input manually.
4. **Baseline:** Think of a straightforward correct solution, even if it is slow.
5. **Improve:** Look for repeated work, suitable data structures, or a known pattern.
6. **Prove:** Explain why the approach gives the correct result.
7. **Analyze:** Estimate time and extra memory.
8. **Test:** Try normal, boundary, duplicate, empty, and unusual inputs as appropriate.
9. **Review:** Can you make the code simpler or easier to maintain?

This process is useful for LeetCode, technical interviews, and everyday programming.

## Example problem: find whether a list contains a target

Input: `numbers = [4, 9, 2, 7]`, `target = 2`  
Output: `True`

A direct approach is to inspect values one by one until the target is found or the list ends.

```python
def contains_target(numbers, target):
    # Visit each value once. Stop early when the target is found.
    for number in numbers:
        if number == target:
            return True
    return False

print(contains_target([4, 9, 2, 7], 2))  # True
print(contains_target([4, 9, 2, 7], 5))  # False
print(contains_target([], 5))             # False
```

**Analysis:** In the worst case, every one of the `n` values is checked, so time complexity is `O(n)`. The function uses only a fixed amount of extra memory, so auxiliary space is `O(1)`.

An empty list has no ordinary arithmetic average. Decide how your program should handle it rather than allowing an unclear failure.

```python
def calculate_average(numbers):
    if not numbers:
        # Explicitly define the behavior for empty input.
        raise ValueError("Cannot calculate the average of an empty list")
    return sum(numbers) / len(numbers)

print(calculate_average([2, 4, 6]))

# Uncomment to test the empty-input behavior:
# calculate_average([])
```

# 3. Dry runs and tracing variables

A **dry run** means manually following the code with a small example. It helps you catch logic errors before execution.

# 4. Brute force versus optimization

**Brute force** is a direct solution that tries the obvious possibilities. It is often easier to reason about and can be a good starting point.

An **optimized solution** reduces repeated work by using a better algorithm, data structure, or mathematical observation.

## Example: does a list contain a duplicate?

A brute-force solution compares every pair. It can take quadratic time, `O(n²)`.

```python
def contains_duplicate_brute_force(numbers):
    n = len(numbers)

    for i in range(n):
        for j in range(i + 1, n):
            if numbers[i] == numbers[j]:
                return True

    return False

print(contains_duplicate_brute_force([4, 1, 7, 1]))
print(contains_duplicate_brute_force([4, 1, 7, 9]))
```

An optimized approach tracks values already seen in a set. Set membership is typically average-case `O(1)`, making the overall expected time `O(n)` at the cost of extra memory.

```python
def contains_duplicate_set(numbers):
    seen = set()

    for number in numbers:
        if number in seen:
            return True
        seen.add(number)

    return False

print(contains_duplicate_set([4, 1, 7, 1]))
print(contains_duplicate_set([4, 1, 7, 9]))
```

**Trade-off:** The set-based method usually uses `O(n)` extra space. The brute-force method uses `O(1)` auxiliary space. A faster algorithm can require more memory, so consider both time and space.

# 5. What is Big-O notation?

Big-O describes an upper-bound growth rate for an algorithm as input size increases. In introductory algorithm analysis, it is commonly used to summarize worst-case growth.

Suppose an algorithm performs about `3n + 10` basic operations. As `n` grows, the linear term dominates. We describe the growth as `O(n)`.

When simplifying Big-O:
- Ignore constant multipliers: `O(3n)` becomes `O(n)`.
- Ignore lower-order terms: `O(n² + n)` becomes `O(n²)`.
- Focus on how the resource requirement scales with input size.

Big-O does not tell you the exact runtime in seconds. Hardware, language implementation, input data, and constant factors still matter.

## Big-O versus Theta and Omega

- `O(f(n))`: an asymptotic upper bound.
- `Ω(f(n))`: an asymptotic lower bound.
- `Θ(f(n))`: a tight asymptotic bound, when both upper and lower bounds match.

Many beginner explanations use Big-O to discuss worst-case complexity. That is a useful convention, but technically Big-O itself does not automatically mean worst case.

# 6. Common complexity classes

From slower-growing to faster-growing for sufficiently large `n`:

| Complexity | Name | Example |
|---|---|---|
| `O(1)` | Constant | Access a list element by index |
| `O(log n)` | Logarithmic | Binary search on sorted data |
| `O(n)` | Linear | Scan a list once |
| `O(n log n)` | Linearithmic | Typical efficient comparison sorts |
| `O(n²)` | Quadratic | Compare every pair |
| `O(n³)` | Cubic | Three nested loops over `n` |
| `O(2ⁿ)` | Exponential | Some exhaustive subset searches |
| `O(n!)` | Factorial | Enumerate all permutations |

This is a general guide. Actual complexity depends on the exact operation and algorithm.

# 7. Count operations in loops

## 7.1 One loop: `O(n)`

If a loop processes each of `n` items once, it performs a number of iterations proportional to `n`.

```python
def print_items(items):
    for item in items:
        print(item)

# The loop runs once per item: O(n) time.
print_items([10, 20, 30])
```

## 7.2 Two separate loops: still `O(n)`

If one loop runs `n` times and another loop runs `n` times, the total is `n + n = 2n`. Drop the constant factor: `O(n)`.

```python
def two_passes(numbers):
    # First pass.
    for number in numbers:
        _ = number

    # Second pass.
    for number in numbers:
        _ = number

# Time: O(n) + O(n) = O(n).
two_passes([1, 2, 3])
```

## 7.3 Nested loops: `O(n²)`

If each of `n` outer-loop iterations runs an inner loop `n` times, there are `n × n` inner iterations.

```python
def print_all_pairs(numbers):
    for first in numbers:
        for second in numbers:
            print(first, second)

# With n items, this prints n^2 pairs.
print_all_pairs([1, 2, 3])
```

## 7.4 Dependent nested loops can be different

The inner loop does not always run `n` times. In this example, it runs fewer times on each successive outer iteration. The total is `n(n-1)/2`, which is still `O(n²)`.

```python
def print_unique_index_pairs(numbers):
    n = len(numbers)
    for i in range(n):
        for j in range(i + 1, n):
            print(i, j)

print_unique_index_pairs([10, 20, 30, 40])
```

# 8. Loops that halve or double

A loop that repeatedly halves a value often runs `O(log n)` times because the input size shrinks by a constant factor each iteration.

```python
def halve_until_one(n):
    steps = 0

    while n > 1:
        n //= 2
        steps += 1

    return steps

for value in [8, 16, 32, 100]:
    print(value, "->", halve_until_one(value), "steps")
```

For example, repeatedly halving 32 gives `32 → 16 → 8 → 4 → 2 → 1`: only five steps. Doubling a counter until it reaches `n` also takes logarithmically many iterations.

Remember: `log₂(n)` and `log₁₀(n)` differ by a constant factor, so their asymptotic complexity is both `O(log n)`.

# 10. Space complexity

Space complexity describes how memory requirements grow with input size.

Distinguish:
- **Input space:** memory occupied by the input itself.
- **Auxiliary space:** additional memory used by the algorithm.
- **Total space:** may include both input and auxiliary memory, depending on the convention being used.

## Example A: constant auxiliary space

```python
def sum_values(numbers):
    total = 0  # One accumulator, regardless of list length.

    for number in numbers:
        total += number

    return total

# Time: O(n). Auxiliary space: O(1).
print(sum_values([1, 2, 3, 4]))
```

## Example B: linear auxiliary space

```python
def make_squares(numbers):
    result = []

    for number in numbers:
        result.append(number * number)

    return result

# Time: O(n). Auxiliary output space: O(n).
print(make_squares([1, 2, 3, 4]))
```

## Example C: recursion uses call-stack space

A recursive function may use memory for each active call. A recursion depth proportional to `n` usually implies `O(n)` call-stack space.

```python
def recursive_sum(n):
    if n <= 0:
        return 0
    return n + recursive_sum(n - 1)

print(recursive_sum(5))
# Time: O(n). Call-stack space: O(n).
```

# 11. Best, average, and worst cases

Consider linear search:
- **Best case:** target is first, so `O(1)`.
- **Worst case:** target is last or absent, so `O(n)`.
- **Average case:** under common assumptions about target position, a linear number of items are inspected, so `O(n)`.

Always say which case you are analyzing when it matters. The assumptions behind an average-case analysis should be clear.

```python
def linear_search(numbers, target):
    for index, number in enumerate(numbers):
        if number == target:
            return index
    return -1

data = [5, 9, 2, 7]

print(linear_search(data, 5))   # Best-case example: index 0
print(linear_search(data, 7))   # Later match
print(linear_search(data, 100)) # Worst-case example: target absent
```

# 12. Correctness and loop invariants

Efficiency matters only if the solution is correct.

A **loop invariant** is a statement that remains true at a particular point in each loop iteration. For example, when summing a list, after processing the first `k` items, `total` equals the sum of those `k` items.

A simple correctness argument often has three parts:
1. **Initialization:** the statement is true before the first iteration.
2. **Maintenance:** if it is true before an iteration, the loop preserves it.
3. **Termination:** when the loop ends, the invariant helps establish the desired result.

```python
def sum_values_with_trace(numbers):
    total = 0

    # Invariant: total equals the sum of all items processed so far.
    for number in numbers:
        total += number
        print("Processed", number, "| total so far =", total)

    return total

print("Final result:", sum_values_with_trace([2, 4, 6]))
```

# 13. Useful Python problem-solving patterns

These patterns will appear repeatedly in LeetCode problems. Learn to recognize what each one is useful for.

## 13.1 Running minimum or maximum

Use a variable to track the best value seen so far. This often gives `O(n)` time and `O(1)` auxiliary space.

```python
def find_largest(numbers):
    if not numbers:
        raise ValueError("numbers must not be empty")

    largest = numbers[0]

    for number in numbers[1:]:
        if number > largest:
            largest = number

    return largest

print(find_largest([-5, 12, 3, 8]))
```

## 13.2 Early return

If you have enough information to answer the question, return immediately. This can improve practical runtime, although the worst-case complexity may remain unchanged.

```python
def has_negative(numbers):
    for number in numbers:
        if number < 0:
            return True
    return False

print(has_negative([2, 4, -1, 8]))
```

# 14. Frequency counting and hash-based lookup

Frequency counting is a common technique for problems about duplicates, anagrams, counts, and pairs.

A dictionary maps each value to the number of times it has appeared.

```python
def count_frequencies(numbers):
    counts = {}

    for number in numbers:
        counts[number] = counts.get(number, 0) + 1

    return counts

print(count_frequencies([2, 1, 2, 3, 1, 2]))
```

## Example: Two Sum using a dictionary

Given a list and a target, return indices of two different elements whose values add to the target. This version returns the first pair found, or `None` if no pair exists.

```python
def two_sum(numbers, target):
    # Map a value to the index where it was first seen.
    seen = {}

    for index, number in enumerate(numbers):
        needed = target - number

        # If the complement appeared earlier, we have a pair.
        if needed in seen:
            return [seen[needed], index]

        # Store the current value only after checking its complement.
        seen[number] = index

    return None

print(two_sum([2, 7, 11, 15], 9))  # [0, 1]
print(two_sum([3, 3], 6))          # [0, 1]
print(two_sum([1, 2, 3], 10))      # None
```

**Complexity:** Expected `O(n)` time and `O(n)` extra space, because each value is processed once and the dictionary can grow with the input.

The order of checking and storing matters: it prevents using the same element twice in the same pair.

# 15. Prefix sums

A prefix sum stores cumulative totals. It can answer range-sum queries efficiently after preprocessing.

For an array `numbers`, define `prefix[i]` as the sum of the first `i` elements. Then the sum from index `left` through `right`, inclusive, is `prefix[right + 1] - prefix[left]`.

```python
def build_prefix_sums(numbers):
    prefix = [0]

    for number in numbers:
        prefix.append(prefix[-1] + number)

    return prefix

numbers = [3, 1, 4, 2]
prefix = build_prefix_sums(numbers)

print(prefix)  # [0, 3, 4, 8, 10]

left, right = 1, 3
range_sum = prefix[right + 1] - prefix[left]
print(range_sum)  # 1 + 4 + 2 = 7
```

**Complexity:** Building prefix sums takes `O(n)` time and `O(n)` extra space. Each subsequent range-sum query takes `O(1)` time.

# 16. Two pointers

Two pointers means maintaining two indexes or references that move through data. It is especially useful with sorted arrays or sequences.

## Example: check whether a sorted list has a pair summing to a target

If the sum is too small, move the left pointer right. If the sum is too large, move the right pointer left.

```python
def has_pair_with_sum_sorted(numbers, target):
    left = 0
    right = len(numbers) - 1

    while left < right:
        current_sum = numbers[left] + numbers[right]

        if current_sum == target:
            return True
        elif current_sum < target:
            left += 1
        else:
            right -= 1

    return False

print(has_pair_with_sum_sorted([1, 2, 4, 6, 9], 10))  # True
print(has_pair_with_sum_sorted([1, 2, 4, 6, 9], 20))  # False
```

**Important:** The input must be sorted for this pointer movement rule to be valid. Sorting first costs `O(n log n)` time, unless the input is already sorted or the sorting cost is handled elsewhere.

For an already sorted list, this scan is `O(n)` time and `O(1)` auxiliary space.

# 17. Sliding window

A sliding window maintains a moving range of elements. It is useful for contiguous subarray or substring problems.

## Example: maximum sum of any contiguous subarray of fixed length `k`

```python
def max_sum_subarray(numbers, k):
    if k <= 0 or k > len(numbers):
        raise ValueError("k must be between 1 and len(numbers)")

    # Compute the first window's sum.
    window_sum = sum(numbers[:k])
    best_sum = window_sum

    # Slide the window one position at a time.
    for right in range(k, len(numbers)):
        window_sum += numbers[right]       # Add the new rightmost item.
        window_sum -= numbers[right - k]   # Remove the item leaving the window.
        best_sum = max(best_sum, window_sum)

    return best_sum

print(max_sum_subarray([2, 1, 5, 1, 3, 2], 3))  # 9
```

**Complexity:** `O(n)` time and `O(1)` auxiliary space for the fixed-size window. The first `sum(numbers[:k])` creates a slice of length `k`, but that initialization is bounded by the input length; for a strict constant-auxiliary-space implementation, compute the initial sum with a loop instead of creating a slice.

# 18. Binary search

Binary search works on sorted data. Each comparison discards roughly half of the remaining search interval, so the time complexity is `O(log n)`.

```python
def binary_search(numbers, target):
    left = 0
    right = len(numbers) - 1

    while left <= right:
        middle = (left + right) // 2

        if numbers[middle] == target:
            return middle
        elif numbers[middle] < target:
            left = middle + 1
        else:
            right = middle - 1

    return -1

sorted_numbers = [2, 5, 8, 12, 16, 23, 38]
print(binary_search(sorted_numbers, 16))  # Index 4
print(binary_search(sorted_numbers, 7))   # -1
```

**Precondition:** the list must be sorted in ascending order for this implementation to work correctly.

- Time: `O(log n)`.
- Auxiliary space: `O(1)` for this iterative version.
- A recursive version generally uses `O(log n)` call-stack space.

# 19. Sorting and searching trade-offs

Sorting can make later operations easier, but sorting has a cost.

Example: checking whether a list has duplicates:
- Use a set for expected `O(n)` time and `O(n)` extra space.
- Sort a copy and compare neighbors: typically `O(n log n)` time and extra memory for the copy.
- Sort the original list in place: typically `O(n log n)` time, but it changes the input.

Always check whether the problem allows changing the input and whether original order matters.

```python
def has_duplicate_by_sorting(numbers):
    ordered = sorted(numbers)  # Creates a sorted copy.

    for index in range(1, len(ordered)):
        if ordered[index] == ordered[index - 1]:
            return True

    return False

print(has_duplicate_by_sorting([4, 2, 7, 2]))
print(has_duplicate_by_sorting([4, 2, 7, 9]))
```

# 20. Recursion and recurrence intuition

For recursive algorithms, consider:
1. How many recursive calls happen?
2. How much work happens in each call?
3. How does the input size change?
4. How deep can the call stack become?

For example, `recursive_sum(n)` calls itself on `n - 1` and does constant extra work per call. It therefore takes `O(n)` time and `O(n)` stack space.

Do not assume every recursive algorithm is exponential. The complexity depends on the number and size of recursive calls.

```python
def factorial_recursive(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n <= 1:
        return 1
    return n * factorial_recursive(n - 1)

print(factorial_recursive(5))
```

## Fibonacci: a warning about repeated work

The naive recursive Fibonacci function recalculates the same values many times, giving exponential time. Memoization avoids repeating solved subproblems.

```python
def fibonacci_naive(n):
    if n <= 1:
        return n
    return fibonacci_naive(n - 1) + fibonacci_naive(n - 2)

def fibonacci_memoized(n, memo=None):
    if n <= 1:
        return n
    if memo is None:
        memo = {}
    if n not in memo:
        memo[n] = fibonacci_memoized(n - 1, memo) + fibonacci_memoized(n - 2, memo)
    return memo[n]

print(fibonacci_naive(10))
print(fibonacci_memoized(40))
```

For the naive Fibonacci implementation, the time grows exponentially (commonly bounded as `O(2ⁿ)`). The memoized version computes each subproblem once, giving `O(n)` time and `O(n)` memory for the memo and recursion stack.

# 23. Solutions with explanations

```python
# 1. Minimum in a non-empty list
def minimum_value(numbers):
    if not numbers:
        raise ValueError("numbers must not be empty")

    smallest = numbers[0]
    for number in numbers[1:]:
        if number < smallest:
            smallest = number
    return smallest

print(minimum_value([3, -1, 8, 2]))
# Time O(n), auxiliary space O(1).
```

```python
# 2. Duplicate detection with a set
def has_duplicate(numbers):
    seen = set()
    for number in numbers:
        if number in seen:
            return True
        seen.add(number)
    return False

print(has_duplicate([1, 2, 1]))
print(has_duplicate([1, 2, 3]))
# Expected time O(n), auxiliary space O(n).
```

```python
# 3. Character frequency
def character_frequency(text):
    counts = {}
    for character in text:
        counts[character] = counts.get(character, 0) + 1
    return counts

print(character_frequency("aab"))
# Time O(n), auxiliary space O(k), where k is the number of distinct characters.
```

```python
# 4. Two Sum using a hash map
def two_sum_indices(numbers, target):
    seen = {}
    for index, number in enumerate(numbers):
        needed = target - number
        if needed in seen:
            return [seen[needed], index]
        seen[number] = index
    return None

print(two_sum_indices([2, 7, 11, 15], 9))
# Expected time O(n), auxiliary space O(n).
```

```python
# 5. Range sum using prefix sums
def range_sum(numbers, left, right):
    prefix = [0]
    for number in numbers:
        prefix.append(prefix[-1] + number)
    return prefix[right + 1] - prefix[left]

print(range_sum([3, 1, 4, 2], 1, 3))
# Preprocessing time O(n), auxiliary space O(n); each query O(1) after preprocessing.
```

## Additional practice explanations

**6. Two pointers:** On an already sorted list, initialize pointers at both ends. Move the left pointer when the sum is too small and the right pointer when the sum is too large. Time `O(n)`, auxiliary space `O(1)`.

**7. Fixed-size sliding window:** Compute the first window sum, then subtract the outgoing item and add the incoming item for each move. Time `O(n)`, auxiliary space `O(1)` if the initial window is summed without making a slice.

**8. Binary search:** Compare the middle value and discard half of the remaining search range. Time `O(log n)`, auxiliary space `O(1)` for the iterative implementation.

**9. Triple nested loops:** Each of the three loops contributes a factor of `n`, so the total iteration count is proportional to `n × n × n = n³`, giving `O(n³)`.

**10. Set versus sorting:** A set-based check has expected `O(n)` time and `O(n)` extra space. Sorting is typically `O(n log n)` time; sorting a copy requires extra storage, while in-place sorting changes the input.

# 24. Big-O quick reference

| Code shape / technique | Typical time | Typical auxiliary space |
|---|---:|---:|
| Fixed number of operations | `O(1)` | `O(1)` |
| One full scan | `O(n)` | `O(1)` if no growing structure |
| Two separate full scans | `O(n)` | Depends on stored data |
| Two nested full scans | `O(n²)` | `O(1)` if only counters |
| Three nested full scans | `O(n³)` | `O(1)` if only counters |
| Repeatedly halve search range | `O(log n)` | `O(1)` iterative |
| Efficient comparison sort | `O(n log n)` | Depends on algorithm |
| Hash map / set scan | Expected `O(n)` | Often `O(n)` |
| Prefix sum preprocessing | `O(n)` | `O(n)` |
| Two pointers on sorted data | `O(n)` | `O(1)` |
| Fixed-size sliding window | `O(n)` | `O(1)` |
| Binary search | `O(log n)` | `O(1)` iterative |
| Naive recursive Fibonacci | Exponential | Recursion stack grows with `n` |
