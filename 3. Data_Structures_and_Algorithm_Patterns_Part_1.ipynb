# 1. Choosing a data structure

A data structure is a way to organize information so operations can be performed effectively.

| Need | Common choice | Why |
|---|---|---|
| Access by position | List / array | Indexing is typically `O(1)` |
| Fast average membership checks | Set | Hash-based lookup is typically average `O(1)` |
| Count occurrences or map keys to values | Dictionary | Fast average lookup and update |
| Last-in, first-out behavior | Stack | The newest item is removed first |
| First-in, first-out behavior | Queue / deque | The oldest item is removed first |
| Insert/delete nodes without shifting an array | Linked list | Local pointer changes can be `O(1)` when node is known |
| Repeated minimum/maximum extraction | Heap | Efficient access to one extreme |
| Hierarchical relationships | Tree | Represents parent-child structure |
| General connections between entities | Graph | Represents arbitrary relationships |

There is no universally best structure. Choose based on the operations the problem needs.

## 2.3 In-place reversal with two pointers

Swap the left and right items, then move inward. This takes `O(n)` time and `O(1)` auxiliary space.

```python
def reverse_in_place(items):
    left = 0
    right = len(items) - 1

    while left < right:
        # Swap the values at the two ends.
        items[left], items[right] = items[right], items[left]
        left += 1
        right -= 1

    return items

print(reverse_in_place([1, 2, 3, 4, 5]))
print(reverse_in_place([]))
```

# 3. Array patterns

## 3.1 Running best: maximum subarray sum (Kadane's algorithm)

Find the maximum sum of a **non-empty contiguous subarray**. At each position, either:
- extend the previous subarray, or
- start a new subarray at the current number.

This works with negative values too, unlike initializing the answer to zero.

```python
def max_subarray_sum(numbers):
    if not numbers:
        raise ValueError("Input must contain at least one number")

    current_sum = numbers[0]
    best_sum = numbers[0]

    for number in numbers[1:]:
        # Either continue the previous subarray or start fresh.
        current_sum = max(number, current_sum + number)
        best_sum = max(best_sum, current_sum)

    return best_sum

print(max_subarray_sum([-2, 1, -3, 4, -1, 2, 1, -5, 4]))  # 6
print(max_subarray_sum([-8, -3, -6]))                       # -3
```

**Complexity:** `O(n)` time and `O(1)` auxiliary space.

## 3.2 Move zeros to the end

Keep a write pointer for the next position that should receive a non-zero value. This version preserves the relative order of non-zero values.

```python
def move_zeros_to_end(numbers):
    write = 0

    # Copy non-zero values toward the front.
    for read in range(len(numbers)):
        if numbers[read] != 0:
            numbers[write], numbers[read] = numbers[read], numbers[write]
            write += 1

    return numbers

print(move_zeros_to_end([0, 1, 0, 3, 12]))
```

**Complexity:** `O(n)` time and `O(1)` auxiliary space, because the list is modified in place.

# 4. Two pointers

Two pointers maintain two indexes that move through a sequence. The technique often reduces a nested-loop solution to a single pass.

Common uses:
- Pair sums in sorted arrays.
- Reversing a sequence.
- Palindrome checks.
- Removing duplicates from sorted arrays.
- Comparing elements from both ends.

## 4.1 Palindrome check

Ignore non-alphanumeric characters and letter case. Move inward while comparing the next valid characters.

```python
def is_palindrome(text):
    left = 0
    right = len(text) - 1

    while left < right:
        # Skip characters that are not letters or digits.
        while left < right and not text[left].isalnum():
            left += 1
        while left < right and not text[right].isalnum():
            right -= 1

        if text[left].lower() != text[right].lower():
            return False

        left += 1
        right -= 1

    return True

print(is_palindrome("A man, a plan, a canal: Panama"))  # True
print(is_palindrome("Python"))                         # False
```

## 4.2 Remove duplicates from a sorted array

Because equal values are adjacent in a sorted list, a write pointer can retain one copy of each distinct value.

```python
def remove_duplicates_sorted(numbers):
    if not numbers:
        return 0

    write = 1

    for read in range(1, len(numbers)):
        if numbers[read] != numbers[write - 1]:
            numbers[write] = numbers[read]
            write += 1

    return write

data = [1, 1, 2, 2, 2, 3, 4, 4]
length = remove_duplicates_sorted(data)
print("Unique length:", length)
print("Unique prefix:", data[:length])
```

**Complexity:** `O(n)` time and `O(1)` auxiliary space. The input must be sorted for this particular method.

# 5. Sliding window

A sliding window tracks a contiguous section of a list or string.

Use it when the problem mentions:
- A contiguous subarray or substring.
- A fixed window length `k`.
- The longest or shortest window satisfying a condition.
- A moving range where adjacent windows overlap heavily.

## 5.1 Fixed-size window

Find the maximum sum of any `k` consecutive values. Update the sum by adding the new value and subtracting the value that leaves the window.

```python
def max_sum_fixed_window(numbers, k):
    if k <= 0 or k > len(numbers):
        raise ValueError("k must be between 1 and len(numbers)")

    # Build the first window sum without making a slice.
    window_sum = 0
    for index in range(k):
        window_sum += numbers[index]

    best = window_sum

    for right in range(k, len(numbers)):
        window_sum += numbers[right]       # New item enters the window.
        window_sum -= numbers[right - k]   # Old item leaves the window.
        best = max(best, window_sum)

    return best

print(max_sum_fixed_window([2, 1, 5, 1, 3, 2], 3))  # 9
```

## 5.2 Variable-size window: longest substring without repeated characters

Expand the right pointer. If a repeated character appears, move the left boundary until the current window has no duplicate characters.

```python
def longest_unique_substring_length(text):
    last_seen = {}
    left = 0
    best = 0

    for right, character in enumerate(text):
        # If this character is still inside the current window, move left.
        if character in last_seen and last_seen[character] >= left:
            left = last_seen[character] + 1

        last_seen[character] = right
        best = max(best, right - left + 1)

    return best

print(longest_unique_substring_length("abcabcbb"))  # 3
print(longest_unique_substring_length("bbbbb"))     # 1
print(longest_unique_substring_length(""))           # 0
```

**Complexity:** `O(n)` expected time and `O(k)` space, where `k` is the number of distinct characters tracked (at most the number of characters in the input).

**Important distinction:** Sliding windows generally apply to contiguous ranges. A two-pointer pair-sum technique does not require the values to be contiguous.

# 6. Prefix sums and difference arrays

## 6.1 Prefix sums

Prefix sums store cumulative totals so that many range-sum queries can be answered quickly.

```python
def build_prefix(numbers):
    prefix = [0]

    for number in numbers:
        prefix.append(prefix[-1] + number)

    return prefix

numbers = [3, 1, 4, 2]
prefix = build_prefix(numbers)
print(prefix)  # [0, 3, 4, 8, 10]

# Sum from left=1 through right=3, inclusive:
left, right = 1, 3
print(prefix[right + 1] - prefix[left])  # 7
```

Building the prefix list takes `O(n)` time and `O(n)` space. Each range query then takes `O(1)` time.

## 6.2 Difference arrays

A difference array is useful when applying many range updates. To add `delta` to each element from index `left` through `right`, record `+delta` at `left` and `-delta` just after `right`, then compute a running sum at the end.

```python
def apply_range_updates(length, updates):
    # Difference array stores where each update begins and ends.
    difference = [0] * (length + 1)

    for left, right, delta in updates:
        difference[left] += delta
        difference[right + 1] -= delta

    result = []
    running = 0

    for index in range(length):
        running += difference[index]
        result.append(running)

    return result

updates = [(1, 3, 2), (2, 4, 3)]
print(apply_range_updates(5, updates))  # [0, 2, 5, 5, 3]
```

**Complexity:** `O(n + q)` time for an array of length `n` and `q` updates, and `O(n)` extra space.

# 7. Strings and character patterns

Strings are immutable sequences. Common interview operations include counting characters, checking anagrams, reversing text, and processing substrings.

## 7.1 Anagram check with frequency counts

Two strings are anagrams when they contain the same characters with the same frequencies. This example ignores spaces and letter case.

```python
def are_anagrams(first, second):
    first = "".join(character.lower() for character in first if character.isalnum())
    second = "".join(character.lower() for character in second if character.isalnum())

    if len(first) != len(second):
        return False

    counts = {}

    for character in first:
        counts[character] = counts.get(character, 0) + 1

    for character in second:
        if character not in counts:
            return False
        counts[character] -= 1
        if counts[character] < 0:
            return False

    return True

print(are_anagrams("Listen", "Silent"))       # True
print(are_anagrams("hello", "world"))         # False
print(are_anagrams("Dormitory", "Dirty room")) # True
```

This method takes `O(n + m)` expected time and `O(k)` auxiliary space for `k` distinct characters, excluding normalized strings that are created as part of preprocessing. A sorting approach is also common and typically takes `O(n log n + m log m)` time.

## 7.2 Group words that are anagrams

A canonical key can be the sorted letters of a word. Words with the same key belong in the same group.

```python
def group_anagrams(words):
    groups = {}

    for word in words:
        key = tuple(sorted(word.lower()))
        groups.setdefault(key, []).append(word)

    return list(groups.values())

print(group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"]))
```

The sorting-based method typically takes `O(w * m log m)` time for `w` words of average length `m`, ignoring details of output construction.

# 8. Hash maps and sets

A **set** tracks distinct values. A **dictionary** maps keys to values. Both are hash-based in Python, with typical average-case `O(1)` membership, insertion, and lookup.

## 8.1 Two Sum

For each number, calculate its complement (`target - number`). If the complement has already been seen, return both indexes.

```python
def two_sum(numbers, target):
    seen = {}

    for index, number in enumerate(numbers):
        complement = target - number

        if complement in seen:
            return [seen[complement], index]

        # Store after checking, so one element is not used twice.
        seen[number] = index

    return None

print(two_sum([2, 7, 11, 15], 9))  # [0, 1]
print(two_sum([3, 3], 6))          # [0, 1]
```

Expected time: `O(n)`. Extra space: `O(n)`.

## 8.2 First non-repeating character

```python
def first_unique_character(text):
    counts = {}

    for character in text:
        counts[character] = counts.get(character, 0) + 1

    for index, character in enumerate(text):
        if counts[character] == 1:
            return index

    return -1

print(first_unique_character("leetcode"))  # 0
print(first_unique_character("aabb"))      # -1
```

## 9.1 Valid parentheses

Push opening brackets. When a closing bracket appears, it must match the most recent unmatched opening bracket.

```python
def is_valid_parentheses(text):
    stack = []
    matching_open = {")": "(", "]": "[", "}": "{"}

    for character in text:
        if character in "([{":
            stack.append(character)
        elif character in matching_open:
            if not stack or stack.pop() != matching_open[character]:
                return False

    return len(stack) == 0

print(is_valid_parentheses("()[]{}"))  # True
print(is_valid_parentheses("([)]"))    # False
print(is_valid_parentheses("((("))     # False
```

**Complexity:** `O(n)` time and up to `O(n)` stack space.

# 10. Monotonic stacks

A monotonic stack keeps values in increasing or decreasing order. It is useful for “next greater element,” “next smaller element,” and histogram problems.

## 10.1 Next greater element to the right

For each item, find the first greater value to its right. Keep indexes of values that have not found their next greater value yet. When a larger value arrives, resolve those indexes.

```python
def next_greater_elements(numbers):
    result = [-1] * len(numbers)
    stack = []  # Stores indexes whose next greater item is not found.

    for index, number in enumerate(numbers):
        while stack and numbers[stack[-1]] < number:
            previous_index = stack.pop()
            result[previous_index] = number

        stack.append(index)

    return result

print(next_greater_elements([2, 1, 2, 4, 3]))  # [4, 2, 4, -1, -1]
```

Although there is a nested `while` loop, each index is pushed once and popped at most once. Therefore the total time is `O(n)`, not `O(n²)`. The stack and result use `O(n)` space.

# 11. Queues and deques

A queue follows **FIFO**: first in, first out. Avoid repeatedly using `list.pop(0)` for a queue because removing the first list element shifts the remaining elements.

Use `collections.deque`, which supports efficient appends and pops at both ends.

```python
from collections import deque

queue = deque()
queue.append("first")
queue.append("second")
queue.append("third")

print(queue.popleft())  # first
print(queue.popleft())  # second
print(queue)
```

## 11.1 Breadth-first traversal of a simple graph

BFS uses a queue to explore neighbors level by level. Graphs will be covered in more depth later; this example introduces the queue pattern.

```python
from collections import deque

def bfs(graph, start):
    visited = {start}
    queue = deque([start])
    order = []

    while queue:
        node = queue.popleft()
        order.append(node)

        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return order

graph = {
    "A": ["B", "C"],
    "B": ["D"],
    "C": ["E"],
    "D": [],
    "E": []
}

print(bfs(graph, "A"))  # One valid order: A, B, C, D, E
```

For a graph with `V` vertices and `E` edges, BFS takes `O(V + E)` time and `O(V)` extra space, assuming an adjacency-list representation.

# 12. Linked lists

A linked list is made of nodes. Each node stores a value and a reference to the next node. Unlike an array, nodes are not necessarily stored next to one another in memory.

In Python, we can create a simple singly linked list for learning. LeetCode often provides a `ListNode` class for linked-list problems.

```python
class ListNode:
    def __init__(self, value=0, next_node=None):
        self.value = value
        self.next = next_node

# Build: 10 -> 20 -> 30
head = ListNode(10, ListNode(20, ListNode(30)))

current = head
while current is not None:
    print(current.value)
    current = current.next
```

## 12.1 Reverse a linked list

Use three references: `previous`, `current`, and `next_node`. Reverse one pointer at a time so you do not lose the rest of the list.

```python
def reverse_linked_list(head):
    previous = None
    current = head

    while current is not None:
        next_node = current.next  # Save the rest of the list.
        current.next = previous   # Reverse the current pointer.
        previous = current        # Move previous forward.
        current = next_node       # Move current forward.

    return previous

def linked_list_to_list(head):
    values = []
    current = head
    while current is not None:
        values.append(current.value)
        current = current.next
    return values

head = ListNode(1, ListNode(2, ListNode(3)))
print(linked_list_to_list(reverse_linked_list(head)))  # [3, 2, 1]
```

**Complexity:** Reversal takes `O(n)` time and `O(1)` auxiliary space.

## 12.2 Fast and slow pointers: cycle detection

Move one pointer by one node and another by two nodes. If they meet, a cycle exists. If the fast pointer reaches the end, there is no cycle.

```python
def has_cycle(head):
    slow = head
    fast = head

    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next

        if slow is fast:
            return True

    return False

# A list without a cycle.
a = ListNode(1, ListNode(2, ListNode(3)))
print(has_cycle(a))  # False

# Create a cycle: node 3 points back to node 2.
first = ListNode(1)
second = ListNode(2)
third = ListNode(3)
first.next = second
second.next = third
third.next = second
print(has_cycle(first))  # True
```

**Complexity:** `O(n)` time and `O(1)` auxiliary space. Use `is` to compare node identity; two separate nodes can have equal values without being the same node.

## 13.1 Merge two sorted lists

Two pointers can merge two sorted lists in linear time.

```python
def merge_sorted_lists(first, second):
    i = 0
    j = 0
    merged = []

    while i < len(first) and j < len(second):
        if first[i] <= second[j]:
            merged.append(first[i])
            i += 1
        else:
            merged.append(second[j])
            j += 1

    # Append any remaining values.
    merged.extend(first[i:])
    merged.extend(second[j:])

    return merged

print(merge_sorted_lists([1, 3, 5], [2, 4, 6, 8]))
```

**Complexity:** `O(n + m)` time and `O(n + m)` output space for lists of lengths `n` and `m`.

# 14. Binary search patterns

Binary search requires a monotonic condition or sorted data. Small mistakes in boundaries are common, so choose a clear template and test it carefully.

## 14.1 Exact search

```python
def binary_search(numbers, target):
    left = 0
    right = len(numbers) - 1

    while left <= right:
        middle = left + (right - left) // 2

        if numbers[middle] == target:
            return middle
        elif numbers[middle] < target:
            left = middle + 1
        else:
            right = middle - 1

    return -1

print(binary_search([1, 3, 5, 7, 9], 7))
print(binary_search([1, 3, 5, 7, 9], 4))
```

## 14.2 Lower bound: first index whose value is at least target

This template returns an index from `0` through `len(numbers)`. If every item is smaller than the target, it returns `len(numbers)`.

```python
def lower_bound(numbers, target):
    left = 0
    right = len(numbers)  # Half-open interval [left, right).

    while left < right:
        middle = (left + right) // 2

        if numbers[middle] < target:
            left = middle + 1
        else:
            right = middle

    return left

data = [1, 2, 2, 2, 5, 8]
print(lower_bound(data, 2))  # 1
print(lower_bound(data, 3))  # 4
print(lower_bound(data, 9))  # 6
```

## 14.3 Search on the answer

Sometimes the answer is a number and we can test whether a candidate is feasible. If feasibility is monotonic (for example, all values at or above a threshold work), binary search can find the boundary.

Before using this technique, prove that the condition changes in only one direction.

```python
def first_true(low, high, condition):
    # Find the first value in [low, high] where condition(value) is True.
    # Assumption: condition is monotonic: False...False, True...True.
    answer = high + 1

    while low <= high:
        middle = (low + high) // 2

        if condition(middle):
            answer = middle
            high = middle - 1
        else:
            low = middle + 1

    return answer

# First integer from 1 through 100 whose square is at least 30.
print(first_true(1, 100, lambda x: x * x >= 30))  # 6
```

# 15. Intervals

Interval problems involve ranges such as `[start, end]`. A common strategy is to sort by start time, then merge overlapping intervals.

```python
def merge_intervals(intervals):
    if not intervals:
        return []

    intervals = sorted(intervals, key=lambda interval: interval[0])
    merged = [intervals[0][:]]  # Copy the first interval.

    for start, end in intervals[1:]:
        last_end = merged[-1][1]

        if start <= last_end:
            # Intervals overlap; extend the last merged interval.
            merged[-1][1] = max(last_end, end)
        else:
            merged.append([start, end])

    return merged

print(merge_intervals([[1, 3], [2, 6], [8, 10], [15, 18]]))
```

**Complexity:** Sorting costs `O(n log n)` and the merge pass costs `O(n)`, so total time is `O(n log n)`. Output storage can be `O(n)`.

Check the problem's definition of overlap: some problems treat touching endpoints as overlapping, while others do not.

# 16. Matrix traversal

A matrix is often represented as a list of lists. Be careful with row and column indexes, and check whether the matrix is empty before reading `matrix[0]`.

```python
def matrix_row_sums(matrix):
    # Return the sum of each row.
    return [sum(row) for row in matrix]

def matrix_column_sums(matrix):
    if not matrix:
        return []

    rows = len(matrix)
    columns = len(matrix[0])
    result = []

    for column in range(columns):
        total = 0
        for row in range(rows):
            total += matrix[row][column]
        result.append(total)

    return result

matrix = [
    [1, 2, 3],
    [4, 5, 6]
]
print(matrix_row_sums(matrix))     # [6, 15]
print(matrix_column_sums(matrix))  # [5, 7, 9]
```

A complete matrix traversal generally takes `O(rows × columns)` time. For ragged lists (rows with different lengths), column-based indexing needs extra care.

# 17. Pattern selection guide

| Problem clue | Pattern to consider | Important condition |
|---|---|---|
| Find a pair in a sorted array | Two pointers | Input is sorted |
| Longest/shortest contiguous segment | Sliding window | Window condition can be maintained as it moves |
| Many range-sum queries | Prefix sums | Array values do not change between queries, unless updates are handled |
| Count duplicates or occurrences | Dictionary / Counter | Values can be used as hashable keys |
| Matching brackets or next greater item | Stack | Order of unresolved items matters |
| First-in, first-out processing | Queue / deque | Process items in arrival order |
| Repeatedly find minimum or maximum | Heap | Need repeated extreme extraction |
| Sorted search or monotonic feasibility | Binary search | Ordering/monotonicity is guaranteed |
| Merge overlapping ranges | Sort + interval scan | Interval endpoint rules are clear |
| Reverse or detect cycle in linked list | Pointer techniques | Follow node references carefully |
| Explore graph level by level | BFS + queue | Track visited vertices |

```python
# 1. Count occurrences
def count_occurrences(numbers, target):
    count = 0
    for number in numbers:
        if number == target:
            count += 1
    return count

print(count_occurrences([1, 2, 2, 3, 2], 2))  # 3
```

```python
# 2. Move zeros to the end, preserving non-zero order
def move_zeros(numbers):
    write = 0
    for read in range(len(numbers)):
        if numbers[read] != 0:
            numbers[write], numbers[read] = numbers[read], numbers[write]
            write += 1
    return numbers

print(move_zeros([0, 1, 0, 3, 12]))
```

```python
# 3. Maximum sum of a non-empty contiguous subarray
def max_subarray(numbers):
    if not numbers:
        raise ValueError("Input must not be empty")
    current = best = numbers[0]
    for number in numbers[1:]:
        current = max(number, current + number)
        best = max(best, current)
    return best

print(max_subarray([-2, 1, -3, 4, -1, 2, 1, -5, 4]))  # 6
```

```python
# 4. Phrase palindrome
def phrase_is_palindrome(text):
    cleaned = "".join(ch.lower() for ch in text if ch.isalnum())
    return cleaned == cleaned[::-1]

print(phrase_is_palindrome("A man, a plan, a canal: Panama"))
```

```python
# 5. Anagrams, ignoring spaces and case
def anagrams(first, second):
    first = "".join(ch.lower() for ch in first if not ch.isspace())
    second = "".join(ch.lower() for ch in second if not ch.isspace())
    if len(first) != len(second):
        return False
    counts = {}
    for ch in first:
        counts[ch] = counts.get(ch, 0) + 1
    for ch in second:
        if counts.get(ch, 0) == 0:
            return False
        counts[ch] -= 1
    return True

print(anagrams("Dormitory", "Dirty room"))
```

```python
# 6. Longest substring without repeated characters
def longest_unique(text):
    last_seen = {}
    left = best = 0
    for right, ch in enumerate(text):
        if ch in last_seen and last_seen[ch] >= left:
            left = last_seen[ch] + 1
        last_seen[ch] = right
        best = max(best, right - left + 1)
    return best

print(longest_unique("pwwkew"))  # 3
```

```python
# 7. Two Sum
def two_sum(numbers, target):
    seen = {}
    for i, value in enumerate(numbers):
        need = target - value
        if need in seen:
            return [seen[need], i]
        seen[value] = i
    return None

print(two_sum([2, 7, 11, 15], 9))
```

```python
# 8. Valid brackets
def valid_brackets(text):
    stack = []
    pairs = {")": "(", "]": "[", "}": "{"}
    for ch in text:
        if ch in "([{":
            stack.append(ch)
        elif ch in pairs:
            if not stack or stack.pop() != pairs[ch]:
                return False
    return not stack

print(valid_brackets("{[()]}"))
```

```python
# 9. Next greater element
def next_greater(numbers):
    answer = [-1] * len(numbers)
    stack = []
    for i, value in enumerate(numbers):
        while stack and numbers[stack[-1]] < value:
            answer[stack.pop()] = value
        stack.append(i)
    return answer

print(next_greater([2, 1, 2, 4, 3]))
```

```python
# 10. Reverse linked list
def reverse_list(head):
    previous = None
    current = head
    while current:
        next_node = current.next
        current.next = previous
        previous = current
        current = next_node
    return previous

head = ListNode(1, ListNode(2, ListNode(3)))
print(linked_list_to_list(reverse_list(head)))
```

```python
# 11. Lower bound
def lower_bound(numbers, target):
    left, right = 0, len(numbers)
    while left < right:
        middle = (left + right) // 2
        if numbers[middle] < target:
            left = middle + 1
        else:
            right = middle
    return left

print(lower_bound([1, 2, 2, 4, 7], 2))  # 1
```

```python
# 12. Merge intervals
def merge_intervals(intervals):
    if not intervals:
        return []
    intervals = sorted(intervals, key=lambda pair: pair[0])
    result = [intervals[0][:]]
    for start, end in intervals[1:]:
        if start <= result[-1][1]:
            result[-1][1] = max(result[-1][1], end)
        else:
            result.append([start, end])
    return result

print(merge_intervals([[1, 4], [2, 5], [8, 9]]))
```
