# 1. Recursion review

Recursion is when a function calls itself to solve a smaller version of the same problem.

A correct recursive function needs:
- **Base case:** when to stop.
- **Progress:** each call must move toward a base case.
- **Combination:** how smaller results produce the current result.

Every active call uses call-stack space. Deep recursion may hit Python's recursion limit, so an iterative approach may be safer for very deep inputs.

```python
def factorial(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n <= 1:                 # Base case
        return 1
    return n * factorial(n - 1)  # Smaller subproblem

print(factorial(5))  # 120
```

## 1.1 Recursion versus iteration

A recursive solution can be elegant, but iteration may use less stack space. Compare both versions of summing numbers from 1 through `n`.

```python
def sum_recursive(n):
    if n <= 0:
        return 0
    return n + sum_recursive(n - 1)

def sum_iterative(n):
    total = 0
    for value in range(1, n + 1):
        total += value
    return total

print(sum_recursive(5))
print(sum_iterative(5))
```

Both take `O(n)` time. The recursive version uses `O(n)` call-stack space; the iterative version uses `O(1)` auxiliary space.

# 2. Backtracking

Backtracking explores a choice, recursively explores the result, and then undoes the choice to try another possibility.

General template:
1. Check whether the current state is a complete solution.
2. Iterate through possible choices.
3. Make a choice.
4. Recurse.
5. Undo the choice.

Backtracking can explore exponentially many possibilities. Pruning invalid partial solutions early can reduce the search.

## 2.1 Generate all subsets

```python
def subsets(numbers):
    result = []
    current = []

    def backtrack(index):
        # Every position has been considered: record this subset.
        if index == len(numbers):
            result.append(current.copy())
            return

        # Choice 1: exclude numbers[index].
        backtrack(index + 1)

        # Choice 2: include numbers[index].
        current.append(numbers[index])
        backtrack(index + 1)

        # Undo the include choice before returning.
        current.pop()

    backtrack(0)
    return result

print(subsets([1, 2, 3]))
```

There are `2ⁿ` subsets, so output generation takes at least `O(2ⁿ)` time; copying each subset adds work, commonly summarized as `O(n · 2ⁿ)` time. The output itself stores `O(n · 2ⁿ)` elements. The recursion depth is `O(n)`.

## 2.2 Generate permutations

```python
def permutations(numbers):
    result = []
    current = []
    used = [False] * len(numbers)

    def backtrack():
        if len(current) == len(numbers):
            result.append(current.copy())
            return

        for index, number in enumerate(numbers):
            if used[index]:
                continue

            used[index] = True
            current.append(number)

            backtrack()

            # Undo the choice.
            current.pop()
            used[index] = False

    backtrack()
    return result

print(permutations([1, 2, 3]))
```

There are `n!` permutations. Materializing them takes `O(n · n!)` time and output space. If the input contains duplicates and unique permutations are required, add a duplicate-skipping rule after sorting the input.

# 3. Binary trees

A tree consists of nodes connected by parent-child relationships. In a binary tree, each node has at most two children: `left` and `right`.

Common traversals:
- **Preorder:** node, left, right.
- **Inorder:** left, node, right.
- **Postorder:** left, right, node.
- **Level order:** visit nodes level by level using a queue.

```python
class TreeNode:
    def __init__(self, value=0, left=None, right=None):
        self.value = value
        self.left = left
        self.right = right

# Build this tree:
#       4
#      / \
#     2   6
#    / \   \
#   1   3   7
root = TreeNode(
    4,
    TreeNode(2, TreeNode(1), TreeNode(3)),
    TreeNode(6, None, TreeNode(7))
)
```

## 3.1 Recursive depth-first traversals

```python
def preorder(root):
    result = []

    def visit(node):
        if node is None:
            return
        result.append(node.value)
        visit(node.left)
        visit(node.right)

    visit(root)
    return result

def inorder(root):
    result = []

    def visit(node):
        if node is None:
            return
        visit(node.left)
        result.append(node.value)
        visit(node.right)

    visit(root)
    return result

def postorder(root):
    result = []

    def visit(node):
        if node is None:
            return
        visit(node.left)
        visit(node.right)
        result.append(node.value)

    visit(root)
    return result

print("Preorder:", preorder(root))
print("Inorder:", inorder(root))
print("Postorder:", postorder(root))
```

Each traversal visits each node once: `O(n)` time. Recursive auxiliary space is `O(h)`, where `h` is tree height. A balanced tree has height `O(log n)`, while a highly skewed tree can have height `O(n)`.

## 3.2 Level-order traversal (BFS)

```python
from collections import deque

def level_order(root):
    if root is None:
        return []

    result = []
    queue = deque([root])

    while queue:
        node = queue.popleft()
        result.append(node.value)

        if node.left is not None:
            queue.append(node.left)
        if node.right is not None:
            queue.append(node.right)

    return result

print(level_order(root))  # [4, 2, 6, 1, 3, 7]
```

## 3.3 Tree height and node count

```python
def tree_height(root):
    # Height measured in nodes: empty tree has height 0.
    if root is None:
        return 0
    return 1 + max(tree_height(root.left), tree_height(root.right))

def count_nodes(root):
    if root is None:
        return 0
    return 1 + count_nodes(root.left) + count_nodes(root.right)

print("Height:", tree_height(root))
print("Nodes:", count_nodes(root))
```

# 4. Binary search trees (BSTs)

A binary search tree typically follows:
- Values in the left subtree are smaller than the node.
- Values in the right subtree are larger than the node.

Duplicate-value policies vary by problem. The following examples assume distinct values.

Search, insertion, and deletion take `O(h)`, where `h` is tree height. This is `O(log n)` for a balanced tree, but can degrade to `O(n)` for a skewed tree.

```python
def bst_search(root, target):
    current = root

    while current is not None:
        if current.value == target:
            return current
        elif target < current.value:
            current = current.left
        else:
            current = current.right

    return None

print(bst_search(root, 3).value)
print(bst_search(root, 99))
```

## 4.1 Validate a BST using bounds

Checking only each node against its immediate children is insufficient. A node must satisfy all ancestor bounds.

```python
def is_valid_bst(root):
    def validate(node, low, high):
        if node is None:
            return True

        if not (low < node.value < high):
            return False

        return (
            validate(node.left, low, node.value)
            and validate(node.right, node.value, high)
        )

    return validate(root, float("-inf"), float("inf"))

print(is_valid_bst(root))  # True
```

# 5. Tree problem patterns

## 5.1 Maximum depth

Maximum depth is the number of nodes on the longest root-to-leaf path.

```python
def max_depth(root):
    if root is None:
        return 0
    return 1 + max(max_depth(root.left), max_depth(root.right))

print(max_depth(root))
```

## 5.2 Invert a binary tree

Swap the left and right subtrees at every node.

```python
def invert_tree(root):
    if root is None:
        return None

    root.left, root.right = root.right, root.left
    invert_tree(root.left)
    invert_tree(root.right)
    return root

# Use a separate tree if you need to preserve the original.
small_tree = TreeNode(2, TreeNode(1), TreeNode(3))
print(level_order(invert_tree(small_tree)))  # [2, 3, 1]
```

## 5.3 Lowest common ancestor in a BST

For a BST, if both targets are smaller than the current node, go left. If both are larger, go right. Otherwise, the current node is the split point.

```python
def lowest_common_ancestor_bst(root, p, q):
    current = root

    while current is not None:
        if p < current.value and q < current.value:
            current = current.left
        elif p > current.value and q > current.value:
            current = current.right
        else:
            return current

    return None

print(lowest_common_ancestor_bst(root, 1, 3).value)  # 2
```

# 6. Heaps and priority queues

A heap supports efficient access to the smallest item (min-heap) or largest item (max-heap). Python's `heapq` implements a min-heap.

Typical costs:
- Peek at minimum: `O(1)`.
- Push: `O(log n)`.
- Pop minimum: `O(log n)`.
- Build a heap from a list: `O(n)` with `heapify()`.

Use heaps for top-k values, scheduling, merging sorted streams, and algorithms that repeatedly need the next smallest item.

```python
import heapq

values = [7, 2, 9, 1, 5]
heapq.heapify(values)  # Build a heap in place.

print("Minimum:", values[0])
heapq.heappush(values, 3)
print("Remove minimum:", heapq.heappop(values))
print("Heap storage:", values)  # Internal heap order is not fully sorted.
```

## 6.1 K largest values

Maintain a min-heap of size `k`. If it grows beyond `k`, remove its smallest item. The heap then retains the largest `k` values seen so far.

```python
def k_largest(numbers, k):
    if k <= 0:
        return []
    heap = []

    for number in numbers:
        heapq.heappush(heap, number)
        if len(heap) > k:
            heapq.heappop(heap)

    return sorted(heap, reverse=True)

print(k_largest([4, 1, 9, 7, 2, 8], 3))  # [9, 8, 7]
```

Time is `O(n log k)` for maintaining the heap, plus `O(k log k)` to sort the final values. Extra space is `O(k)`.

## 6.2 Merge k sorted lists

A heap stores the current smallest candidate from each list. After removing one item, push the next item from that same list.

```python
def merge_k_sorted_lists(sorted_lists):
    heap = []

    for list_index, values in enumerate(sorted_lists):
        if values:
            heapq.heappush(heap, (values[0], list_index, 0))

    result = []

    while heap:
        value, list_index, item_index = heapq.heappop(heap)
        result.append(value)

        next_index = item_index + 1
        if next_index < len(sorted_lists[list_index]):
            next_value = sorted_lists[list_index][next_index]
            heapq.heappush(heap, (next_value, list_index, next_index))

    return result

print(merge_k_sorted_lists([[1, 4, 7], [2, 5], [3, 6, 8]]))
```

For `N` total values across `k` lists, time is `O(N log k)` and heap space is `O(k)`.

# 7. Graph representations

A graph contains vertices (nodes) and edges (connections). It can be directed or undirected, weighted or unweighted.

An adjacency list stores each vertex's neighbors. It is often a good choice for sparse graphs.

```python
# Undirected graph: every edge is stored in both directions.
graph = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "D"],
    "D": ["B", "C"]
}

for node, neighbors in graph.items():
    print(node, "->", neighbors)
```

An adjacency list uses `O(V + E)` space. An adjacency matrix uses `O(V²)` space but can check whether an edge exists in `O(1)` time.

# 8. Graph BFS and DFS

## 8.1 Breadth-first search (BFS)

BFS explores nodes in layers. In an unweighted graph, BFS finds shortest paths by number of edges.

```python
from collections import deque

def graph_bfs(graph, start):
    visited = {start}
    queue = deque([start])
    order = []

    while queue:
        node = queue.popleft()
        order.append(node)

        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                visited.add(neighbor)  # Mark when enqueued to avoid duplicates.
                queue.append(neighbor)

    return order

print(graph_bfs(graph, "A"))
```

## 8.2 Depth-first search (DFS)

DFS explores one path as far as possible before backtracking. It can be implemented recursively or with an explicit stack.

```python
def graph_dfs(graph, start):
    visited = set()
    order = []

    def visit(node):
        if node in visited:
            return

        visited.add(node)
        order.append(node)

        for neighbor in graph.get(node, []):
            visit(neighbor)

    visit(start)
    return order

print(graph_dfs(graph, "A"))
```

For an adjacency-list graph, BFS and DFS take `O(V + E)` time and `O(V)` auxiliary space. The exact traversal order depends on neighbor order.

# 9. Cycle detection and topological sorting

## 9.1 Detect a cycle in an undirected graph

When exploring an undirected graph, encountering an already-visited neighbor other than the parent indicates a cycle.

```python
def has_undirected_cycle(graph):
    visited = set()

    def dfs(node, parent):
        visited.add(node)

        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                if dfs(neighbor, node):
                    return True
            elif neighbor != parent:
                return True

        return False

    for node in graph:
        if node not in visited and dfs(node, None):
            return True

    return False

print(has_undirected_cycle({"A": ["B"], "B": ["A", "C"], "C": ["B"]}))  # False
print(has_undirected_cycle({"A": ["B", "C"], "B": ["A", "C"], "C": ["A", "B"]}))  # True
```

## 9.2 Topological sort using Kahn's algorithm

A topological ordering exists only for a directed acyclic graph (DAG). Kahn's algorithm repeatedly removes nodes with indegree zero.

```python
from collections import deque

def topological_sort(num_nodes, edges):
    graph = [[] for _ in range(num_nodes)]
    indegree = [0] * num_nodes

    for source, destination in edges:
        graph[source].append(destination)
        indegree[destination] += 1

    queue = deque(
        node for node in range(num_nodes)
        if indegree[node] == 0
    )
    order = []

    while queue:
        node = queue.popleft()
        order.append(node)

        for neighbor in graph[node]:
            indegree[neighbor] -= 1
            if indegree[neighbor] == 0:
                queue.append(neighbor)

    # If some nodes remain, the graph contains a directed cycle.
    if len(order) != num_nodes:
        return None

    return order

edges = [(0, 2), (1, 2), (2, 3)]
print(topological_sort(4, edges))
print(topological_sort(2, [(0, 1), (1, 0)]))  # None: cycle
```

Time complexity is `O(V + E)`, and auxiliary space is `O(V + E)` for the graph, indegrees, queue, and output.

# 10. Shortest paths

## 10.1 BFS for unweighted graphs

If every edge has the same cost, BFS finds the path with the fewest edges. Track distance as each node is first discovered.

```python
def shortest_unweighted_distances(graph, start):
    distance = {start: 0}
    queue = deque([start])

    while queue:
        node = queue.popleft()

        for neighbor in graph.get(node, []):
            if neighbor not in distance:
                distance[neighbor] = distance[node] + 1
                queue.append(neighbor)

    return distance

print(shortest_unweighted_distances(graph, "A"))
```

## 10.2 Dijkstra's algorithm for non-negative edge weights

Dijkstra uses a min-heap to repeatedly process the known closest node. It requires **non-negative edge weights**. It is not correct for graphs with negative-weight edges.

```python
def dijkstra(weighted_graph, start):
    # weighted_graph[node] contains (neighbor, non_negative_weight) pairs.
    distances = {start: 0}
    heap = [(0, start)]

    while heap:
        current_distance, node = heapq.heappop(heap)

        # Ignore stale heap entries.
        if current_distance != distances.get(node, float("inf")):
            continue

        for neighbor, weight in weighted_graph.get(node, []):
            if weight < 0:
                raise ValueError("Dijkstra requires non-negative edge weights")

            candidate = current_distance + weight
            if candidate < distances.get(neighbor, float("inf")):
                distances[neighbor] = candidate
                heapq.heappush(heap, (candidate, neighbor))

    return distances

weighted_graph = {
    "A": [("B", 4), ("C", 1)],
    "C": [("B", 2), ("D", 5)],
    "B": [("D", 1)],
    "D": []
}
print(dijkstra(weighted_graph, "A"))
```

With an adjacency list and binary heap, the common bound is `O((V + E) log V)` (often simplified to `O(E log V)` for connected sparse graphs). Distances and heap entries require additional memory.

For negative edge weights, other algorithms such as Bellman–Ford may be appropriate.

# 11. Union-find / Disjoint Set Union (DSU)

Union-find maintains groups of connected items. It supports:
- `find(x)`: identify the representative of x's component.
- `union(a, b)`: merge the components containing a and b.
- `connected(a, b)`: check whether two values share a component.

Path compression and union by size/rank make operations nearly constant amortized time.

```python
class UnionFind:
    def __init__(self, size):
        self.parent = list(range(size))
        self.size = [1] * size

    def find(self, item):
        # Path compression: point nodes closer to the root.
        if self.parent[item] != item:
            self.parent[item] = self.find(self.parent[item])
        return self.parent[item]

    def union(self, first, second):
        root_first = self.find(first)
        root_second = self.find(second)

        if root_first == root_second:
            return False  # Already in the same component.

        # Attach the smaller component under the larger one.
        if self.size[root_first] < self.size[root_second]:
            root_first, root_second = root_second, root_first

        self.parent[root_second] = root_first
        self.size[root_first] += self.size[root_second]
        return True

    def connected(self, first, second):
        return self.find(first) == self.find(second)

uf = UnionFind(5)
uf.union(0, 1)
uf.union(1, 2)
print(uf.connected(0, 2))  # True
print(uf.connected(0, 4))  # False
```

Union-find is useful for connectivity queries, detecting cycles while processing undirected edges, and minimum spanning tree algorithms such as Kruskal's. With path compression and union by size, operations have amortized near-constant complexity, commonly expressed using the inverse Ackermann function.

# 12. Greedy algorithms

A greedy algorithm makes the best-looking local choice at each step. Greedy is correct only when the problem has a property that supports those choices; a locally attractive choice is not automatically globally optimal.

## 12.1 Interval scheduling

Choose the maximum number of non-overlapping intervals by sorting by finishing time and always choosing the next interval that finishes earliest.

```python
def max_non_overlapping_intervals(intervals):
    # Intervals are (start, end). This version allows a new interval
    # to start exactly when the previous one ends.
    intervals = sorted(intervals, key=lambda interval: interval[1])
    chosen = []
    last_end = float("-inf")

    for start, end in intervals:
        if start >= last_end:
            chosen.append((start, end))
            last_end = end

    return chosen

meetings = [(1, 3), (2, 4), (3, 5), (6, 8), (5, 7)]
print(max_non_overlapping_intervals(meetings))
```

Sorting costs `O(n log n)`; the scan costs `O(n)`. Total time is `O(n log n)`.

The correctness intuition is an exchange argument: choosing the interval that ends earliest leaves at least as much remaining time as any other first choice.

## 12.2 When greedy fails

For coin values `[1, 3, 4]` and amount `6`, a greedy “take the largest coin possible” strategy chooses `4 + 1 + 1` (three coins). The optimal solution is `3 + 3` (two coins). A counterexample like this shows why greedy needs justification.

```python
def greedy_coin_count(coins, amount):
    count = 0
    for coin in sorted(coins, reverse=True):
        take = amount // coin
        count += take
        amount -= take * coin
    return count if amount == 0 else None

print(greedy_coin_count([1, 3, 4], 6))  # 3, not optimal
```

# 13. Dynamic programming foundations

Dynamic programming (DP) is useful when a problem has:
- **Overlapping subproblems:** the same smaller problem is solved repeatedly.
- **Optimal substructure:** a larger answer can be built from answers to smaller subproblems.

Two common styles:
- **Memoization:** top-down recursion with cached results.
- **Tabulation:** bottom-up table filling.

A reliable DP design process:
1. Define the state in plain language.
2. Write the recurrence or transition.
3. Define base cases.
4. Decide the evaluation order.
5. Determine which state values must be stored.
6. Analyze time and space.

## 13.1 Fibonacci: memoization

```python
from functools import lru_cache

@lru_cache(maxsize=None)
def fibonacci_memo(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n <= 1:
        return n
    return fibonacci_memo(n - 1) + fibonacci_memo(n - 2)

print(fibonacci_memo(30))
```

Each state from `0` through `n` is calculated once, so time is `O(n)` and cache plus recursion stack use `O(n)` space. Python's recursion limit still applies to deep recursive problems.

## 13.2 Fibonacci: bottom-up tabulation

```python
def fibonacci_tabulation(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n <= 1:
        return n

    dp = [0] * (n + 1)
    dp[0], dp[1] = 0, 1

    for index in range(2, n + 1):
        dp[index] = dp[index - 1] + dp[index - 2]

    return dp[n]

print(fibonacci_tabulation(30))
```

This version takes `O(n)` time and `O(n)` space. Since each state depends only on the previous two values, the stored space can be reduced to `O(1)`.

## 13.3 Climbing stairs

If you can climb either one or two steps at a time, the number of ways to reach step `n` is the sum of the ways to reach steps `n-1` and `n-2`.

```python
def climbing_stairs(n):
    if n < 0:
        return 0
    if n <= 1:
        return 1

    one_step_before = 1
    two_steps_before = 1

    for _ in range(2, n + 1):
        current = one_step_before + two_steps_before
        two_steps_before = one_step_before
        one_step_before = current

    return one_step_before

print(climbing_stairs(5))  # 8
```

Time is `O(n)` and auxiliary space is `O(1)`.

## 13.4 House robber

Choose non-adjacent values to maximize the total. At each index, either skip the current house or take its value plus the best answer two positions back.

```python
def house_robber(values):
    two_back = 0
    one_back = 0

    for value in values:
        take = two_back + value
        skip = one_back
        current = max(take, skip)

        two_back = one_back
        one_back = current

    return one_back

print(house_robber([2, 7, 9, 3, 1]))  # 12
```

Time is `O(n)` and auxiliary space is `O(1)`. This implementation assumes taking no houses is allowed, so it returns zero for an all-negative list. Adapt the base cases if the problem requires choosing at least one item.

## 13.5 0/1 knapsack

Each item may be used at most once. Let `dp[c]` be the best value achievable with capacity `c`. Iterate capacities **backward** so the current item is not reused during the same iteration.

```python
def knapsack_01(weights, values, capacity):
    dp = [0] * (capacity + 1)

    for weight, value in zip(weights, values):
        # Backward direction is important for 0/1 knapsack.
        for current_capacity in range(capacity, weight - 1, -1):
            dp[current_capacity] = max(
                dp[current_capacity],
                dp[current_capacity - weight] + value
            )

    return dp[capacity]

print(knapsack_01([2, 3, 4], [4, 5, 6], 5))  # 9
```

For `n` items and capacity `W`, time is `O(nW)` and space is `O(W)`. This is **pseudo-polynomial**, because it depends on the numeric capacity, not only on the number of bits needed to represent it.

# 14. More dynamic-programming patterns

## 14.1 Longest increasing subsequence (LIS), `O(n log n)`

A subsequence keeps relative order but does not need to be contiguous. Maintain the smallest possible ending value for an increasing subsequence of each length. This technique returns the LIS length.

```python
from bisect import bisect_left

def lis_length(numbers):
    tails = []

    for number in numbers:
        position = bisect_left(tails, number)

        if position == len(tails):
            tails.append(number)
        else:
            tails[position] = number

    return len(tails)

print(lis_length([10, 9, 2, 5, 3, 7, 101, 18]))  # 4
```

This method takes `O(n log n)` time and `O(n)` space. The `tails` list is not necessarily an actual subsequence; it stores useful minimal tail values.

# 15. Tries (prefix trees)

A trie stores strings by shared prefixes. It is useful for prefix search, dictionary lookup, and word-search problems.

```python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_word = False

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        node = self.root
        for character in word:
            if character not in node.children:
                node.children[character] = TrieNode()
            node = node.children[character]
        node.is_word = True

    def search(self, word):
        node = self._walk(word)
        return node is not None and node.is_word

    def starts_with(self, prefix):
        return self._walk(prefix) is not None

    def _walk(self, text):
        node = self.root
        for character in text:
            if character not in node.children:
                return None
            node = node.children[character]
        return node

trie = Trie()
trie.insert("apple")
trie.insert("app")
print(trie.search("app"))        # True
print(trie.search("ap"))         # False: prefix, not full word
print(trie.starts_with("ap"))   # True
```

If a word has length `L`, insertion and lookup take `O(L)` expected time with dictionary children. Space depends on the total number of characters stored and how much prefix sharing occurs.

# 16. Bit manipulation

Integers can be manipulated at the bit level. These operators are useful in subset masks, flags, and selected algorithm problems.

- `&`: bitwise AND
- `|`: bitwise OR
- `^`: bitwise XOR
- `~`: bitwise NOT
- `<<`: left shift
- `>>`: right shift

Python integers have arbitrary precision, so machine-word assumptions from some languages do not always apply directly.

```python
a = 6   # Binary: 110
b = 3   # Binary: 011

print(a & b)   # 2  (010)
print(a | b)   # 7  (111)
print(a ^ b)   # 5  (101)
print(a << 1)  # 12
print(a >> 1)  # 3

# Check whether a non-negative integer is odd.
number = 13
print(number & 1)  # 1 means odd
```

## 16.1 Check, set, clear, and toggle a bit

Bit positions are counted from zero at the rightmost bit.

```python
number = 0b1010  # Decimal 10
position = 1

# Check whether the bit is set.
is_set = (number & (1 << position)) != 0

# Set the bit.
set_result = number | (1 << position)

# Clear the bit.
clear_result = number & ~(1 << position)

# Toggle the bit.
toggle_result = number ^ (1 << position)

print(is_set, bin(set_result), bin(clear_result), bin(toggle_result))
```

## 16.2 Find the unique value when all others occur twice

XOR has useful properties: `x ^ x == 0` and `x ^ 0 == x`. XOR-ing all values cancels pairs.

```python
def single_number(numbers):
    answer = 0

    for number in numbers:
        answer ^= number

    return answer

print(single_number([4, 1, 2, 1, 2]))  # 4
```

Time is `O(n)` and auxiliary space is `O(1)` in the usual algorithmic model. This relies on the problem guarantee that exactly one value occurs once and every other value occurs twice.

# 17. Complexity reference

| Algorithm / structure | Typical time | Auxiliary space |
|---|---:|---:|
| Backtracking subsets | `O(n · 2ⁿ)` including output copying | `O(n)` recursion path, plus output |
| Tree traversal | `O(n)` | `O(h)` recursive stack |
| Heap push/pop | `O(log n)` | `O(1)` extra per operation |
| Top-k with heap | `O(n log k)` | `O(k)` |
| BFS / DFS | `O(V + E)` | `O(V)` |
| Topological sort | `O(V + E)` | `O(V + E)` |
| Dijkstra with heap | `O((V + E) log V)` | `O(V + E)` in typical representation |
| Union-find operations | Amortized near `O(1)` | `O(V)` structure storage |
| Interval scheduling | `O(n log n)` | Depends on sorting/output |

| XOR single number | `O(n)` | `O(1)` |

`n` is input size, `V` vertices, `E` edges, `h` tree height, `k` top-k size, `W` knapsack capacity, and `L` word length.

# 19. Selected solutions

```python
# 1. All subsets
def all_subsets(numbers):
    result = []
    current = []

    def backtrack(index):
        if index == len(numbers):
            result.append(current.copy())
            return

        backtrack(index + 1)
        current.append(numbers[index])
        backtrack(index + 1)
        current.pop()

    backtrack(0)
    return result

print(all_subsets([1, 2, 3]))
```

```python
# 2. All permutations of distinct values
def all_permutations(numbers):
    result = []
    current = []
    used = [False] * len(numbers)

    def backtrack():
        if len(current) == len(numbers):
            result.append(current.copy())
            return
        for i, value in enumerate(numbers):
            if used[i]:
                continue
            used[i] = True
            current.append(value)
            backtrack()
            current.pop()
            used[i] = False

    backtrack()
    return result

print(all_permutations([1, 2, 3]))
```

```python
# 3. Count connected components using DFS
def count_components(num_nodes, edges):
    graph = [[] for _ in range(num_nodes)]
    for a, b in edges:
        graph[a].append(b)
        graph[b].append(a)

    visited = set()
    components = 0

    for node in range(num_nodes):
        if node in visited:
            continue
        components += 1
        stack = [node]
        visited.add(node)

        while stack:
            current = stack.pop()
            for neighbor in graph[current]:
                if neighbor not in visited:
                    visited.add(neighbor)
                    stack.append(neighbor)

    return components

print(count_components(5, [(0, 1), (1, 2), (3, 4)]))  # 2
```

```python
# 4. Minimum coin change using bottom-up DP
def coin_change(coins, amount):
    infinity = amount + 1
    dp = [0] + [infinity] * amount

    for current_amount in range(1, amount + 1):
        for coin in coins:
            if coin <= current_amount:
                dp[current_amount] = min(
                    dp[current_amount],
                    dp[current_amount - coin] + 1
                )

    return -1 if dp[amount] == infinity else dp[amount]

print(coin_change([1, 3, 4], 6))  # 2
print(coin_change([2], 3))        # -1
```

```python
# 5. LIS length using patience-sorting-style tails
from bisect import bisect_left

def lis_length(numbers):
    tails = []
    for value in numbers:
        index = bisect_left(tails, value)
        if index == len(tails):
            tails.append(value)
        else:
            tails[index] = value
    return len(tails)

print(lis_length([10, 9, 2, 5, 3, 7, 101, 18]))  # 4
```
