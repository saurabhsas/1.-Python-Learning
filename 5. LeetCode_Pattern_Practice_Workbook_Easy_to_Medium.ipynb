# Part 1 — Arrays and Hash Maps

## Pattern concept

A hash map (`dict`) is useful when you need to find a value quickly, count occurrences, or remember where an item appeared. A set is useful when you only need to know whether something has appeared before.

Typical expected complexity:
- Dictionary/set lookup: **O(1)** average
- One pass through `n` items: **O(n)**
- Sorting: usually **O(n log n)**

## Problem 1 — Pair Sum (Easy)

Given a list of integers and a target, return the indices of two different values whose sum equals the target. Assume exactly one answer exists.

Example: `nums = [4, 5, 1, 8]`, `target = 9` → indices `[0, 3]`.

**Suggested LeetCode practice:** Two Sum.

### Think before coding
- What information from earlier elements would help?
- Should you store values or indices?
- Why must the current index not be paired with itself?

```python
def pair_sum(nums, target):
    # Store each number's index so far.
    seen = {}

    for i, value in enumerate(nums):
        needed = target - value

        # If the complement appeared earlier, we found the pair.
        if needed in seen:
            return [seen[needed], i]

        # Store the current value only after checking its complement.
        seen[value] = i

    return []  # Defensive fallback if no pair exists.


assert pair_sum([4, 5, 1, 8], 12) == [0, 3]
assert pair_sum([3, 3], 6) == [0, 1]
```

**Complexity:** O(n) average time and O(n) extra space.

**Why it works:** For every current value `x`, a valid partner must be `target - x`. The dictionary tells us whether that partner has already appeared.

## Problem 2 — First Repeated Value (Easy)

Return the first value encountered whose value has appeared earlier. If there is no repeated value, return `None`.

Example: `[7, 2, 4, 2, 7]` → `2`.

**Suggested LeetCode practice:** Contains Duplicate; First Unique Character in a String is a useful related variation.

### Hint
Use a set to remember values already seen.

```python
def first_repeated_value(nums):
    seen = set()

    for value in nums:
        if value in seen:
            return value
        seen.add(value)

    return None


assert first_repeated_value([7, 2, 4, 2, 7]) == 2
assert first_repeated_value([1, 2, 3]) is None
```

## Problem 3 — Frequency Counter (Easy)

Return a dictionary mapping each value to its frequency.

Example: `["a", "b", "a"]` → `{"a": 2, "b": 1}`.

**Suggested LeetCode practice:** Majority Element; Top K Frequent Elements (later, with heaps).

```python
def frequencies(items):
    counts = {}

    for item in items:
        # get(item, 0) returns zero if this item is new.
        counts[item] = counts.get(item, 0) + 1

    return counts


assert frequencies(["a", "b", "a"]) == {"a": 2, "b": 1}
assert frequencies([]) == {}
```

# Part 2 — Two Pointers

## Pattern concept

Two pointers keep track of two positions in a sequence. They are especially useful when:
- the input is sorted;
- you compare both ends;
- you need to modify a list in-place;
- you need to avoid a nested loop.

## Problem 4 — Check Palindrome (Easy)

Return `True` if a string reads the same from both ends. For this exercise, compare characters exactly.

Example: `"level"` → `True`; `"python"` → `False`.

**Suggested LeetCode practice:** Valid Palindrome (official version also includes character filtering).

```python
def is_palindrome(text):
    left, right = 0, len(text) - 1

    while left < right:
        if text[left] != text[right]:
            return False
        left += 1
        right -= 1

    return True


assert is_palindrome("level") is True
assert is_palindrome("python") is False
assert is_palindrome("") is True
```

**Complexity:** O(n) time and O(1) extra space.

## Problem 5 — Two Sum in a Sorted List (Easy)

Given a sorted list of integers, return the two values whose sum is the target. Assume a solution exists.

Example: `[1, 2, 4, 6, 9]`, target `10` → `(1, 9)`.

**Hint:** If the sum is too small, move the left pointer right. If too large, move the right pointer left.

```python
def sorted_pair_sum(nums, target):
    left, right = 0, len(nums) - 1

    while left < right:
        total = nums[left] + nums[right]

        if total == target:
            return (nums[left], nums[right])
        if total < target:
            left += 1
        else:
            right -= 1

    return None


assert sorted_pair_sum([1, 2, 4, 6, 9], 10) == (1, 9)
assert sorted_pair_sum([1, 3, 5], 8) == (3, 5)
```

## Problem 6 — Move Zeros (Easy)

Move every zero to the end while preserving the relative order of nonzero elements. Modify the list in-place.

Example: `[0, 3, 0, 2, 8]` becomes `[3, 2, 8, 0, 0]`.

**Suggested LeetCode practice:** Move Zeroes.

```python
def move_zeros(nums):
    write = 0

    # Copy nonzero values to the earliest available positions.
    for read in range(len(nums)):
        if nums[read] != 0:
            nums[write] = nums[read]
            write += 1

    # Fill the remaining positions with zeroes.
    while write < len(nums):
        nums[write] = 0
        write += 1

    return nums


values = [0, 3, 0, 2, 8]
assert move_zeros(values) == [3, 2, 8, 0, 0]
```

# Part 3 — Sliding Window

## Pattern concept

A sliding window maintains a contiguous section of an array or string. Instead of recalculating every window from scratch, add the new element and remove the element that leaves the window.

Use a fixed-size window when the question asks for exactly `k` consecutive elements. Use a variable-size window when the range expands and contracts based on a condition.

## Problem 7 — Maximum Sum of `k` Consecutive Values (Easy)

Return the largest sum of any `k` consecutive values. Assume `1 <= k <= len(nums)`.

Example: `[2, 1, 5, 1, 3, 2]`, `k = 3` → `9`.

```python
def max_sum_window(nums, k):
    if k < 1 or k > len(nums):
        raise ValueError("k must be between 1 and len(nums)")

    window_sum = sum(nums[:k])
    best = window_sum

    for right in range(k, len(nums)):
        # Remove the value leaving the window and add the new value.
        window_sum += nums[right] - nums[right - k]
        best = max(best, window_sum)

    return best


assert max_sum_window([2, 1, 5, 1, 3, 2], 3) == 9
assert max_sum_window([-4, -2, -7], 2) == -6
```

**Complexity:** O(n) time and O(1) extra space.

## Problem 8 — Longest Substring Without Repeating Characters (Medium)

Given a string, return the length of its longest substring containing no repeated characters.

Example: `"abcabcbb"` → `3`.

**Suggested LeetCode practice:** Longest Substring Without Repeating Characters.

### Hint
Keep a left boundary and a dictionary containing the most recent index for each character. When a repeated character appears inside the current window, move the left boundary past its previous occurrence.

```python
def longest_unique_substring(text):
    last_seen = {}
    left = 0
    best = 0

    for right, char in enumerate(text):
        # Only move left forward; never move it backward.
        if char in last_seen and last_seen[char] >= left:
            left = last_seen[char] + 1

        last_seen[char] = right
        best = max(best, right - left + 1)

    return best


assert longest_unique_substring("abcabcbb") == 3
assert longest_unique_substring("bbbbb") == 1
assert longest_unique_substring("") == 0
assert longest_unique_substring("pwwkew") == 3
```

# Part 4 — Prefix Sums

## Pattern concept

A prefix sum stores the cumulative sum up to each position. It can make range-sum calculations fast.

If `prefix[i]` is the sum of elements before index `i`, then the sum of `nums[left:right+1]` is:

`prefix[right + 1] - prefix[left]`

## Problem 9 — Range Sum Queries (Easy)

Create a helper class that answers inclusive range-sum queries in O(1) time after O(n) preprocessing.

```python
class RangeSum:
    def __init__(self, nums):
        self.prefix = [0]

        for value in nums:
            self.prefix.append(self.prefix[-1] + value)

    def sum_range(self, left, right):
        if left < 0 or right < left or right + 1 >= len(self.prefix):
            raise IndexError("Invalid range")

        return self.prefix[right + 1] - self.prefix[left]


rs = RangeSum([2, -1, 3, 5])
assert rs.sum_range(0, 2) == 4
assert rs.sum_range(1, 3) == 7
```

## Problem 10 — Count Subarrays With Sum `k` (Medium)

Count the contiguous subarrays whose sum equals `k`. Values may be negative, so a simple expanding/shrinking sliding window is not generally safe.

**Suggested LeetCode practice:** Subarray Sum Equals K.

### Key idea
If the current prefix sum is `current`, an earlier prefix sum of `current - k` means the values between those two positions sum to `k`.

```python
def count_subarrays_sum_k(nums, k):
    prefix_count = {0: 1}  # One empty prefix has sum zero.
    current = 0
    answer = 0

    for value in nums:
        current += value
        answer += prefix_count.get(current - k, 0)
        prefix_count[current] = prefix_count.get(current, 0) + 1

    return answer


assert count_subarrays_sum_k([1, 1, 1], 2) == 2
assert count_subarrays_sum_k([1, -1, 0], 0) == 3
```

# Part 5 — Stack and Monotonic Stack

## Pattern concept

A stack follows **last in, first out**. It is useful for matching brackets, undo operations, and problems where the most recent unresolved item matters.

A monotonic stack keeps values in increasing or decreasing order. It helps solve “next greater” or “next smaller” questions in linear time.

## Problem 11 — Valid Parentheses (Easy)

Given a string containing only `()[]{}`, return whether all brackets are correctly matched and nested.

**Suggested LeetCode practice:** Valid Parentheses.

```python
def valid_parentheses(text):
    matching_open = {")": "(", "]": "[", "}": "{"}
    stack = []

    for char in text:
        if char in "([{":
            stack.append(char)
        else:
            if not stack or stack.pop() != matching_open.get(char):
                return False

    return len(stack) == 0


assert valid_parentheses("([]{})") is True
assert valid_parentheses("([)]") is False
assert valid_parentheses("(((") is False
```

## Problem 12 — Next Greater Element to the Right (Medium)

For each value, return the first greater value to its right, or `-1` if none exists.

Example: `[2, 1, 3]` → `[3, 3, -1]`.

```python
def next_greater_elements(nums):
    answer = [-1] * len(nums)
    stack = []  # Store indices whose next-greater value is not known yet.

    for i, value in enumerate(nums):
        while stack and nums[stack[-1]] < value:
            answer[stack.pop()] = value
        stack.append(i)

    return answer


assert next_greater_elements([2, 1, 3]) == [3, 3, -1]
assert next_greater_elements([5, 4, 3]) == [-1, -1, -1]
```

# Part 6 — Binary Search

## Pattern concept

Binary search works on sorted data or a monotonic condition. Each step eliminates about half the remaining search space.

- Search in a sorted list: O(log n) time.
- Search for the first true position in a monotonic condition: also O(log n).

## Problem 13 — Classic Binary Search (Easy)

Return the index of `target` in a sorted list, or `-1` if absent.

**Suggested LeetCode practice:** Binary Search.

```python
def binary_search(nums, target):
    left, right = 0, len(nums) - 1

    while left <= right:
        mid = left + (right - left) // 2

        if nums[mid] == target:
            return mid
        if nums[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1


assert binary_search([-4, 0, 3, 7, 10], 7) == 3
assert binary_search([-4, 0, 3, 7, 10], 8) == -1
```

## Problem 14 — First Position at Least Target (Medium)

Given a sorted list, return the first index whose value is greater than or equal to `target`. If every value is smaller, return `len(nums)`.

This is often called **lower bound**.

```python
def lower_bound(nums, target):
    left, right = 0, len(nums)

    while left < right:
        mid = left + (right - left) // 2

        if nums[mid] < target:
            left = mid + 1
        else:
            right = mid

    return left


assert lower_bound([1, 2, 2, 2, 5], 2) == 1
assert lower_bound([1, 2, 4], 3) == 2
assert lower_bound([1, 2, 4], 9) == 3
```

# Part 7 — Linked Lists

## Pattern concept

A linked list consists of nodes, each holding a value and a reference to the next node. Pointer updates must be done carefully so that nodes are not lost.

## Problem 15 — Reverse a Singly Linked List (Easy)

Reverse the links and return the new head.

**Suggested LeetCode practice:** Reverse Linked List.

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def reverse_list(head):
    previous = None
    current = head

    while current:
        next_node = current.next  # Save the next node before changing the link.
        current.next = previous
        previous = current
        current = next_node

    return previous


def linked_list_values(head):
    values = []
    while head:
        values.append(head.val)
        head = head.next
    return values


head = ListNode(1, ListNode(2, ListNode(3)))
reversed_head = reverse_list(head)
assert linked_list_values(reversed_head) == [3, 2, 1]
```

**Complexity:** O(n) time and O(1) extra space.

## Problem 16 — Detect a Cycle (Easy/Medium)

Return `True` if a linked list contains a cycle.

**Suggested LeetCode practice:** Linked List Cycle.

### Hint
Use slow and fast pointers. If there is a cycle, the faster pointer eventually catches the slower one.

```python
def has_cycle(head):
    slow = fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

        if slow is fast:
            return True

    return False


a, b, c = ListNode(1), ListNode(2), ListNode(3)
a.next, b.next, c.next = b, c, b
assert has_cycle(a) is True
assert has_cycle(ListNode(1)) is False
```

# Part 8 — Trees and Graph Traversal

## Pattern concept

- **DFS** explores deeply before returning.
- **BFS** explores level by level using a queue.
- Tree recursion often asks a question about each node and combines answers from its children.
- Graph traversal must usually track visited nodes to avoid repeated work or infinite loops.

## Problem 17 — Maximum Depth of a Binary Tree (Easy)

Return the number of nodes along the longest path from the root to a leaf.

**Suggested LeetCode practice:** Maximum Depth of Binary Tree.

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def max_depth(root):
    if root is None:
        return 0

    return 1 + max(max_depth(root.left), max_depth(root.right))


tree = TreeNode(1, TreeNode(2, TreeNode(4)), TreeNode(3))
assert max_depth(tree) == 3
assert max_depth(None) == 0
```

## Problem 18 — Level-Order Traversal (Medium)

Return the values grouped by tree level.

**Suggested LeetCode practice:** Binary Tree Level Order Traversal.

```python
from collections import deque


def level_order(root):
    if root is None:
        return []

    result = []
    queue = deque([root])

    while queue:
        level_values = []

        # Process only the nodes that were in the queue at the start of this level.
        for _ in range(len(queue)):
            node = queue.popleft()
            level_values.append(node.val)

            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)

        result.append(level_values)

    return result


tree = TreeNode(1, TreeNode(2), TreeNode(3, TreeNode(4)))
assert level_order(tree) == [[1], [2, 3], [4]]
```

## Problem 19 — Number of Connected Components (Medium)

Given `n` vertices labelled `0` through `n-1` and undirected edges, count connected components.

**Suggested LeetCode practice:** Number of Connected Components in an Undirected Graph (where available) or Number of Provinces.

```python
def count_components(n, edges):
    graph = [[] for _ in range(n)]

    for a, b in edges:
        graph[a].append(b)
        graph[b].append(a)

    visited = set()
    components = 0

    for start in range(n):
        if start in visited:
            continue

        components += 1
        stack = [start]
        visited.add(start)

        while stack:
            node = stack.pop()
            for neighbor in graph[node]:
                if neighbor not in visited:
                    visited.add(neighbor)
                    stack.append(neighbor)

    return components


assert count_components(5, [[0, 1], [1, 2], [3, 4]]) == 2
assert count_components(3, []) == 3
```

# Part 9 — Heap and Top-K

## Pattern concept

A heap efficiently gives access to the smallest or largest-priority item. Python's `heapq` is a min-heap.

A common strategy for **Top K** problems is to maintain a heap of only `k` candidates instead of sorting every item when that is beneficial.

## Problem 20 — K Largest Values (Medium)

Return the `k` largest values in descending order.

```python
import heapq


def k_largest(nums, k):
    if k < 0 or k > len(nums):
        raise ValueError("k must be between 0 and len(nums)")
    if k == 0:
        return []

    # nlargest uses a heap-based strategy when appropriate.
    return heapq.nlargest(k, nums)


assert k_largest([4, 1, 9, 2, 7], 3) == [9, 7, 4]
assert k_largest([4, 1], 0) == []
```

# Part 10 — Backtracking

## Pattern concept

Backtracking builds a candidate solution step by step. It explores a choice, recurses, and then undoes the choice so another branch can be explored.

Common uses: subsets, permutations, combinations, and constraint puzzles.

## Problem 21 — Generate All Subsets (Medium)

Return every subset of a list of distinct values.

**Suggested LeetCode practice:** Subsets.

```python
def subsets(nums):
    result = []
    path = []

    def backtrack(start):
        result.append(path.copy())

        for i in range(start, len(nums)):
            path.append(nums[i])
            backtrack(i + 1)
            path.pop()  # Undo the choice before trying the next value.

    backtrack(0)
    return result


actual = {tuple(sorted(item)) for item in subsets([1, 2, 3])}
expected = {(), (1,), (2,), (3,), (1, 2), (1, 3), (2, 3), (1, 2, 3)}
assert actual == expected
```

**Complexity:** There are `2^n` subsets, so output size alone is exponential.

# Part 11 — Greedy and Intervals

## Pattern concept

A greedy algorithm makes a locally best choice. A greedy solution is not automatically correct: it needs a proof that the local choice can be part of an optimal solution.

## Problem 22 — Maximum Non-Overlapping Intervals (Medium)

Given intervals `[start, end]`, choose the maximum number of mutually non-overlapping intervals. For this exercise, intervals that touch at endpoints are considered non-overlapping.

**Hint:** Sort by end time, then select each interval whose start is at least the previous selected end.

```python
def max_non_overlapping(intervals):
    intervals = sorted(intervals, key=lambda interval: interval[1])
    count = 0
    last_end = float("-inf")

    for start, end in intervals:
        if start >= last_end:
            count += 1
            last_end = end

    return count


assert max_non_overlapping([[1, 3], [2, 4], [3, 5], [6, 8]]) == 3
assert max_non_overlapping([]) == 0
```

# Part 12 — Dynamic Programming

## Pattern concept

Dynamic programming (DP) is useful when a problem has:
1. **Overlapping subproblems:** the same smaller questions recur.
2. **Optimal substructure:** a larger answer can be built from smaller answers.

A practical DP checklist:
1. Define what `dp[i]` means.
2. Write the recurrence.
3. Set base cases.
4. Choose an evaluation order.
5. Return the correct state.
6. Analyze time and space.

## Problem 23 — Climbing Stairs (Easy)

You can climb either one or two steps at a time. Return the number of distinct ways to reach step `n`.

```python
def climbing_stairs(n):
    if n < 0:
        return 0
    if n <= 1:
        return 1

    one_step_before = 1
    two_steps_before = 1

    for _ in range(2, n + 1):
        current = one_step_before + two_steps_before
        two_steps_before = one_step_before
        one_step_before = current

    return one_step_before


assert climbing_stairs(0) == 1
assert climbing_stairs(2) == 2
assert climbing_stairs(5) == 8
```

## Problem 24 — House Robber (Medium)

Each house has a nonnegative amount of money. You cannot rob adjacent houses. Return the maximum amount you can collect.

```python
def house_robber(nums):
    # two_back: best answer before the previous house
    # one_back: best answer through the previous house
    two_back = 0
    one_back = 0

    for amount in nums:
        current = max(one_back, two_back + amount)
        two_back, one_back = one_back, current

    return one_back


assert house_robber([1, 2, 3, 1]) == 4
assert house_robber([2, 7, 9, 3, 1]) == 12
assert house_robber([]) == 0
```

## Problem 25 — Longest Increasing Subsequence (Medium)

Return the length of the longest strictly increasing subsequence. The selected values do not need to be adjacent.

The O(n²) DP approach is a good first solution. A more advanced O(n log n) approach uses binary search.

**Suggested LeetCode practice:** Longest Increasing Subsequence.

```python
def lis_length_quadratic(nums):
    if not nums:
        return 0

    dp = [1] * len(nums)

    for i in range(len(nums)):
        for j in range(i):
            if nums[j] < nums[i]:
                dp[i] = max(dp[i], dp[j] + 1)

    return max(dp)


assert lis_length_quadratic([10, 9, 2, 5, 3, 7, 101, 18]) == 4
assert lis_length_quadratic([3, 2, 1]) == 1
```

3. Answer many range-sum queries on an unchanged array.

## Pattern selection checklist

- Need fast membership/counting? → `dict` or `set`
- Sorted sequence or opposite ends? → two pointers
- Contiguous range? → sliding window or prefix sums
- Nested structure / most recent unresolved item? → stack
- Sorted data or monotonic predicate? → binary search
- Hierarchy? → tree DFS/BFS
- Connections? → graph DFS/BFS/union-find
- All combinations or arrangements? → backtracking
- Best answer built from smaller states? → dynamic programming
- Repeatedly retrieve smallest/largest priority? → heap
