## How to simulate a real interview

For every problem, follow this sequence:

1. **Clarify (2–3 minutes):** Ask about input constraints, duplicates, empty input, and expected output.
2. **Examples (2 minutes):** Create a normal example and one edge case.
3. **Approach (3–5 minutes):** Explain brute force, then the optimization.
4. **Code:** Write the solution without running to the reference solution.
5. **Test:** Trace the example and test boundaries.
6. **Analyze:** State time and auxiliary-space complexity.
7. **Reflect:** Record the first mistake and how you would avoid it next time.

# Round 1 — Easy Warm-up
**Time limit:** 20 minutes total · **Target:** reliable fundamentals

## Problem 1 — First Unique Character (Easy)
Given a string, return the index of its first character that appears exactly once. Return `-1` if no such character exists.

Examples:
- `"leetcode"` → `0`
- `"loveleetcode"` → `2`
- `"aabb"` → `-1`

**Suggested LeetCode practice:** First Unique Character in a String.

### Interview prompts
- What if the string is empty?
- Why is a frequency map useful?
- Can you solve this in two passes?

### Hint ladder — open only if needed

**Hint 1:** Count how often each character appears.

**Hint 2:** Iterate through the string again in its original order.

**Hint 3:** Return the first index whose character count equals one.

```python
from collections import Counter


def first_unique_character(text):
    counts = Counter(text)

    for index, char in enumerate(text):
        if counts[char] == 1:
            return index

    return -1


assert first_unique_character("leetcode") == 0
assert first_unique_character("loveleetcode") == 2
assert first_unique_character("aabb") == -1
assert first_unique_character("") == -1
```

**Reference complexity:** O(n) time and O(k) space, where `k` is the number of distinct characters.

### Self-review
- Did you preserve the original character order?
- Did you test an empty string and a string with no unique character?
- Could you explain why a single frequency-counting pass is not enough by itself to find the first unique index?

# Round 2 — Easy to Medium
**Time limit:** 25 minutes

## Problem 2 — Merge Overlapping Intervals (Medium)
Given a list of intervals, merge intervals that overlap and return the merged intervals sorted by start time.

Example: `[[1,3], [2,6], [8,10], [9,12]]` → `[[1,6], [8,12]]`.

For this exercise, intervals that only touch at an endpoint are considered overlapping if the next start is less than or equal to the current end.

**Suggested LeetCode practice:** Merge Intervals.

### Interview prompts
- Why sort first?
- What invariant does the last interval in the result maintain?
- What is the time complexity?

### Hint ladder

**Hint 1:** Sort intervals by their starting value.

**Hint 2:** Compare the next interval with the last merged interval.

**Hint 3:** If `next_start <= current_end`, extend the current end. Otherwise, append a new interval.

```python
def merge_intervals(intervals):
    if not intervals:
        return []

    ordered = sorted((start, end) for start, end in intervals)
    merged = [list(ordered[0])]

    for start, end in ordered[1:]:
        if start <= merged[-1][1]:
            merged[-1][1] = max(merged[-1][1], end)
        else:
            merged.append([start, end])

    return merged


assert merge_intervals([[1, 3], [2, 6], [8, 10], [9, 12]]) == [[1, 6], [8, 12]]
assert merge_intervals([[1, 4], [4, 5]]) == [[1, 5]]
assert merge_intervals([]) == []
assert merge_intervals([[2, 3]]) == [[2, 3]]
```

**Reference complexity:** O(n log n) time for sorting and O(n) space for the result.

### Self-review
Can you explain why sorting makes it sufficient to compare with only the last merged interval?

# Round 3 — Medium: Sliding Window
**Time limit:** 30 minutes

## Problem 3 — Minimum Window Containing a Target (Hard-style Medium)
Given strings `text` and `target`, return the shortest substring of `text` that contains every character in `target`, including duplicate counts. Return an empty string if no such substring exists.

Example: `text = "ADOBECODEBANC"`, `target = "ABC"` → `"BANC"`.

**Suggested LeetCode practice:** Minimum Window Substring.

### Clarification questions
- Does character order matter? No.
- Do duplicate characters in `target` matter? Yes.
- What if `target` is empty? For this exercise, return `""`.

### Hint ladder

**Hint 1:** Use a frequency counter for the target.

**Hint 2:** Expand the right edge until the current window contains all required character counts.

**Hint 3:** Once valid, move the left edge rightward as far as possible while keeping the window valid.

```python
from collections import Counter, defaultdict


def minimum_window(text, target):
    if not text or not target:
        return ""

    required = Counter(target)
    window = defaultdict(int)
    required_kinds = len(required)
    formed = 0
    left = 0
    best = (float("inf"), 0, 0)

    for right, char in enumerate(text):
        window[char] += 1

        if char in required and window[char] == required[char]:
            formed += 1

        while formed == required_kinds:
            if right - left + 1 < best[0]:
                best = (right - left + 1, left, right + 1)

            left_char = text[left]
            window[left_char] -= 1
            if left_char in required and window[left_char] < required[left_char]:
                formed -= 1
            left += 1

    return "" if best[0] == float("inf") else text[best[1]:best[2]]


assert minimum_window("ADOBECODEBANC", "ABC") == "BANC"
assert minimum_window("a", "a") == "a"
assert minimum_window("a", "aa") == ""
assert minimum_window("aa", "aa") == "aa"
assert minimum_window("abc", "") == ""
```

**Reference complexity:** O(n + m) expected time and O(k) space, where `n` and `m` are string lengths and `k` is the number of distinct characters tracked.

### Common mistakes
- Checking only whether a character exists rather than checking its required count.
- Shrinking the window before it contains all required counts.
- Forgetting to update the validity count when the left character leaves.

# Round 4 — Medium: Binary Search on the Answer
**Time limit:** 30 minutes

## Problem 4 — Minimum Feasible Capacity (Medium)
You are given positive package weights and a number of days. Packages must be shipped in their original order. Find the smallest capacity that allows all packages to be shipped within the given number of days.

Example: `weights = [1,2,3,1,1]`, `days = 4` → `3`.

**Suggested LeetCode practice:** Capacity To Ship Packages Within D Days.

### Interview prompts
- What is the smallest possible capacity?
- What is the largest capacity you ever need?
- Why is feasibility monotonic?

### Hint ladder

**Hint 1:** Search between `max(weights)` and `sum(weights)`.

**Hint 2:** Write a helper that counts the days required for a candidate capacity.

**Hint 3:** If a capacity works, try smaller capacities; otherwise, search larger ones.

```python
def minimum_shipping_capacity(weights, days):
    if not weights or days < 1:
        raise ValueError("weights must be nonempty and days must be positive")

    def feasible(capacity):
        used_days = 1
        load = 0

        for weight in weights:
            if load + weight > capacity:
                used_days += 1
                load = 0
            load += weight

        return used_days <= days

    low, high = max(weights), sum(weights)

    while low < high:
        mid = (low + high) // 2
        if feasible(mid):
            high = mid
        else:
            low = mid + 1

    return low


assert minimum_shipping_capacity([1, 2, 3, 1, 1], 4) == 3
assert minimum_shipping_capacity([3, 2, 2, 4, 1, 4], 3) == 6
assert minimum_shipping_capacity([5], 1) == 5
```

**Reference complexity:** O(n log S) time, where `S = sum(weights)`, and O(1) auxiliary space.

### Self-review
Explain in one sentence why ordinary binary search over the input list is not the right approach here.

# Round 5 — Medium/Hard: Graph Shortest Path
**Time limit:** 35 minutes

## Problem 5 — Network Delay (Medium)
A directed weighted graph has `n` nodes labelled `1` through `n`. Given travel-time edges and a starting node `source`, find the time needed for all nodes to receive the signal. Return `-1` if any node is unreachable.

**Suggested LeetCode practice:** Network Delay Time.

Assume all edge weights are nonnegative.

### Hint ladder

**Hint 1:** You need shortest-path distances from one source.

**Hint 2:** Use Dijkstra's algorithm with a min-heap.

**Hint 3:** The answer is the maximum shortest-path distance, unless some distance is infinite.

```python
import heapq


def network_delay_time(times, n, source):
    graph = [[] for _ in range(n + 1)]
    for u, v, weight in times:
        if weight < 0:
            raise ValueError("Dijkstra requires nonnegative edge weights")
        graph[u].append((v, weight))

    distances = [float("inf")] * (n + 1)
    distances[source] = 0
    heap = [(0, source)]

    while heap:
        distance, node = heapq.heappop(heap)

        if distance != distances[node]:
            continue

        for neighbor, weight in graph[node]:
            candidate = distance + weight
            if candidate < distances[neighbor]:
                distances[neighbor] = candidate
                heapq.heappush(heap, (candidate, neighbor))

    answer = max(distances[1:])
    return -1 if answer == float("inf") else answer


assert network_delay_time([(2, 1, 1), (2, 3, 1), (3, 4, 1)], 4, 2) == 2
assert network_delay_time([(1, 2, 1)], 2, 1) == 1
assert network_delay_time([(1, 2, 1)], 3, 1) == -1
```

**Reference complexity:** O((V + E) log V) time and O(V + E) space.

### Interview follow-up
What changes if negative edge weights are allowed? Explain why Dijkstra's greedy finalization is no longer guaranteed to work and when Bellman–Ford is appropriate.

# Round 6 — Hard: Dynamic Programming
**Time limit:** 40 minutes

## Problem 6 — Longest Increasing Subsequence (Medium/Hard)
Given a list of integers, return the length of the longest strictly increasing subsequence. The selected values need not be adjacent.

**Suggested LeetCode practice:** Longest Increasing Subsequence.

Try the O(n²) dynamic-programming solution first. Then, if time remains, derive the O(n log n) approach using binary search.

### Hint ladder

**Hint 1:** Let `dp[i]` be the longest increasing subsequence ending exactly at index `i`.

**Hint 2:** Consider every earlier index `j` where `nums[j] < nums[i]`.

**Hint 3:** The transition is `dp[i] = 1 + max(dp[j])` over valid `j`, or 1 if no earlier value qualifies.

```python
def lis_length(nums):
    if not nums:
        return 0

    dp = [1] * len(nums)

    for i in range(len(nums)):
        for j in range(i):
            if nums[j] < nums[i]:
                dp[i] = max(dp[i], dp[j] + 1)

    return max(dp)


assert lis_length([10, 9, 2, 5, 3, 7, 101, 18]) == 4
assert lis_length([0, 1, 0, 3, 2, 3]) == 4
assert lis_length([7, 7, 7]) == 1
assert lis_length([]) == 0
```

**Reference complexity:** O(n²) time and O(n) space.

### Advanced follow-up: O(n log n)
Maintain a list `tails`, where `tails[length - 1]` is the smallest possible tail value for an increasing subsequence of that length. For each value, use binary search to replace the first tail that is greater than or equal to it. The length of `tails` is the LIS length. Note that `tails` itself need not be an actual subsequence.

```python
from bisect import bisect_left


def lis_length_nlogn(nums):
    tails = []

    for value in nums:
        position = bisect_left(tails, value)
        if position == len(tails):
            tails.append(value)
        else:
            tails[position] = value

    return len(tails)


assert lis_length_nlogn([10, 9, 2, 5, 3, 7, 101, 18]) == 4
assert lis_length_nlogn([7, 7, 7]) == 1
assert lis_length_nlogn([]) == 0
```

# Round 7 — Hard: Monotonic Stack
**Time limit:** 35 minutes

## Problem 7 — Largest Rectangle in a Histogram (Hard)
Given bar heights, return the area of the largest rectangle that fits within the histogram.

**Suggested LeetCode practice:** Largest Rectangle in Histogram.

### Hint ladder

**Hint 1:** For each bar, you need the nearest shorter bar on both sides.

**Hint 2:** Use a stack of indices whose heights are increasing.

**Hint 3:** When a shorter bar arrives, pop taller bars and calculate their maximum possible width.

```python
def largest_histogram_rectangle(heights):
    stack = []
    best = 0
    values = list(heights) + [0]

    for i, height in enumerate(values):
        while stack and values[stack[-1]] > height:
            popped_height = values[stack.pop()]
            left_boundary = stack[-1] if stack else -1
            width = i - left_boundary - 1
            best = max(best, popped_height * width)
        stack.append(i)

    return best


assert largest_histogram_rectangle([2, 1, 5, 6, 2, 3]) == 10
assert largest_histogram_rectangle([2, 4]) == 4
assert largest_histogram_rectangle([]) == 0
```

**Reference complexity:** O(n) time and O(n) space.

### Debugging prompt
Why do we append a zero-height sentinel? What would happen if we did not process the remaining stack after the loop?

# Round 8 — Hard: Backtracking and Pruning
**Time limit:** 40 minutes

## Problem 8 — N-Queens (Hard)
Place `n` queens on an `n × n` chessboard so no two queens share a row, column, or diagonal. Return the number of valid arrangements.

**Suggested LeetCode practice:** N-Queens.

### Hint ladder

**Hint 1:** Place exactly one queen in each row.

**Hint 2:** Track occupied columns and diagonals.

**Hint 3:** For cell `(row, col)`, diagonal identifiers can be `row - col` and `row + col`.

```python
def count_n_queens(n):
    if n < 1:
        return 0

    columns = set()
    descending_diagonals = set()  # row - col
    ascending_diagonals = set()   # row + col
    solutions = 0

    def backtrack(row):
        nonlocal solutions

        if row == n:
            solutions += 1
            return

        for col in range(n):
            down = row - col
            up = row + col

            if (col in columns or
                down in descending_diagonals or
                up in ascending_diagonals):
                continue

            columns.add(col)
            descending_diagonals.add(down)
            ascending_diagonals.add(up)

            backtrack(row + 1)

            # Undo the choice so the next column can be tried.
            columns.remove(col)
            descending_diagonals.remove(down)
            ascending_diagonals.remove(up)

    backtrack(0)
    return solutions


assert count_n_queens(1) == 1
assert count_n_queens(4) == 2
assert count_n_queens(5) == 10
```

**Complexity:** Exponential in `n` in the worst case; pruning greatly reduces the search compared with checking all board arrangements.

### Interview follow-up
How would you return the actual board arrangements rather than just count them? What additional memory would be required?
